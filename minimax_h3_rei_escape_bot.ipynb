- **合間おな・仰向け** … 走りの合間に床へ仰向け。床は沈まない

**7. 蛾女** — 尾端の第二口が既定

- **蛾女・尾端の第二口（迷ったらこれ）** … 尾先の円口が根元まで密封して蠕動。優先
- **蛾女・上の口（唇と舌）** … 人間の上半身の口。唇と舌の軌道。下半身は虫のまま

**8. 襲う側**

- **レイから襲う（迷ったらこれ）** … レイが腰を掴んで肉壁へ押し付ける
- **サキュバスから襲う** … サキュバスが飛びかかり羽で包む

**9. キス** — しないが既定

- **キス・しない（迷ったらこれ）** … キスカットをスキップ
- **キス・口を開けて舌** … 口を開けて舌が見えるキス。羽と角は残す

**口** — しないが既定。行為は名称ではなく唇と舌の軌道

- **口・しない（迷ったらこれ）** … 口のカットをスキップして体位へ

**体位** — 名称ではなく動き。四つん這い後ろからが既定

- **四つん這い後ろから（迷ったらこれ）** … 掌と膝。後ろから根元まで入れて止めてから往復
