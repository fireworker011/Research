````
def _civitai_api_key():
    import os as _os
    value = _os.environ.get("CIVITAI_API_KEY", "").strip()
    if not value:
        try:
            from google.colab import userdata as _userdata
            value = str(_userdata.get("CIVITAI_API_KEY") or "").strip()
        except Exception:
            value = ""
    if not value:
        raise RuntimeError("CIVITAI_API_KEY が Colab Secrets にも環境変数にもありません。")
    return value
#@title Stable Diffusion web UI reforge（最新版）をGoogle Colabで起動するためのコマンド

#@markdown ## このColabノートブックについて
#@markdown [Stable Diffusion web UI reforge](https://github.com/lllyasviel/stable-diffusion-webui-reForge) をGoogle Colabで起動するためのノートブックとなります。

#@markdown 利用したいモデルや拡張機能、LoRAなどにチェックを入れた上で実行ボタンを押すことで、設定した内容でStable Diffusion web UIが起動します。

#@markdown また各項目には公式ページへのリンクをつけています。利用の際は公式ページよりライセンスなどを確認した上でご利用ください。

### Stable Diffusion web UIインストール
# stable diffusion webuiのリポジトリをクローンして、クローンしたフォルダに移動
!git clone https://github.com/Panchovix/stable-diffusion-webui-reForge.git
%cd /content/stable-diffusion-webui-reForge
!git pull


# 実行時の引数
cmd_args = "--api --share --xformers --theme dark --cuda-malloc --cuda-stream"

#@markdown ---
#@markdown # Civitai API Keyの設定
#@markdown Civitaiにホストされているモデルをダウンロードする場合はAPIキーが必要になりました。APIキーはCivitaiのサイトにログインした状態で以下のリンクから取得できます。

#@markdown https://civitai.com/user/account

#@markdown **なおCivitaiのAPI Keyは人と共有しないようにお気をつけください**

civitai_api_key = "5929d8b685edbeabf15f039561ebe385" #@param {type:"string"}
if not civitai_api_key.strip():
    try:
        civitai_api_key = _civitai_api_key()
    except Exception:
        civitai_api_key = ""

#@markdown # モデル選択
#@markdown 利用したいモデルにチェックを入れてください

#@markdown 生成される画像の傾向を事前に知りたい場合はリンク先のYouTube動画から確認が可能です（作成した動画の宣伝）

#@markdown ---
#@markdown ## 実写系モデル (Base: SD 1.5)

#@markdown ### chilled_remix v2
#@markdown [公式ページ](https://huggingface.co/sazyou-roukaku/chilled_remix)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/c5N1zq6cSNE)
use_chilled_remix_v2 = False #@param {type: "boolean"}

if use_chilled_remix_v2:
  !wget https://huggingface.co/sazyou-roukaku/chilled_remix/resolve/main/chilled_remix_v2.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### chilled_reversemix v2
#@markdown [公式ページ](https://huggingface.co/sazyou-roukaku/chilled_remix)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/c5N1zq6cSNE)
use_chilled_reversemix_v2 = False #@param {type: "boolean"}
if use_chilled_reversemix_v2:
  !wget https://huggingface.co/sazyou-roukaku/chilled_remix/resolve/main/chilled_reversemix_v2.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### BracingEvoMix(v1)
#@markdown [公式ページ](https://huggingface.co/sazyou-roukaku/BracingEvoMix)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/G6mZUq3S354)
use_bracing_evo_mix_v1 = False #@param {type: "boolean"}
if use_bracing_evo_mix_v1:
  !wget https://huggingface.co/sazyou-roukaku/BracingEvoMix/resolve/main/BracingEvoMix_v1.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### BracingEvoMix(v2)
#@markdown [公式ページ](https://huggingface.co/sazyou-roukaku/BracingEvoMix)

use_bracing_evo_mix_v2 = False #@param {type: "boolean"}
if use_bracing_evo_mix_v2:
  !wget https://huggingface.co/sazyou-roukaku/BracingEvoMix/resolve/main/BracingEvoMix_v2.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### BracingEvoMix - Another
#@markdown [公式ページ](https://huggingface.co/sazyou-roukaku/BracingEvoMix)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/hdQxapP2Y98)
use_bracing_evo_another_mix = False #@param {type: "boolean"}
if use_bracing_evo_another_mix:
  !wget https://huggingface.co/sazyou-roukaku/BracingEvoMix/resolve/main/BracingEvoMix_Another_v1.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### BracingEvoMix - Fast
#@markdown [公式ページ](https://huggingface.co/sazyou-roukaku/BracingEvoMix)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/hdQxapP2Y98)
use_bracing_evo_fast_mix = False #@param {type: "boolean"}
if use_bracing_evo_fast_mix:
  !wget https://huggingface.co/sazyou-roukaku/BracingEvoMix/resolve/main/BracingEvoMix_Fast_v1.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### Blessing Mix
#@markdown [公式ページ](https://civitai.com/models/94179/blessing-mix-aka-bracing-evo-mix-clone)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/hNU8CMaewqI)
use_blessing_mix = False #@param {type: "boolean"}
if use_blessing_mix:
  !wget https://civitai.com/api/download/models/100468?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### yayoi_mix v2
#@markdown [公式ページ](https://huggingface.co/Kotajiro/yayoi_mix)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/OM678ZSygow)
use_yayoi_mix_v2 = False #@param {type: "boolean"}
if use_yayoi_mix_v2:
  !wget https://huggingface.co/Kotajiro/yayoi_mix/resolve/main/yayoi_mix_v20-fp16.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### kisaragi_mix v22
#@markdown [公式ページ](https://civitai.com/models/45757/kisaragimix)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/p1zvB-aXKaw)
use_kisaragi_mix_v22 = False #@param {type: "boolean"}
if use_kisaragi_mix_v22:
  !wget https://civitai.com/api/download/models/97186?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### minaduki_mix
#@markdown [公式ページ](https://civitai.com/models/126041/minadukimix)

use_minaduki_mix = False #@param {type: "boolean"}
if use_minaduki_mix:
  !wget https://civitai.com/api/download/models/137751?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### WeddingImperial v1
#@markdown [公式ページ](https://huggingface.co/teasan/WeddingImperial)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/4dukNcNJ_fI)
use_wedding_imperial_v1 = False #@param {type: "boolean"}
if use_wedding_imperial_v1:
  !wget https://huggingface.co/teasan/WeddingImperial/resolve/main/model/weddingImperial.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### WeddingImperial v2
#@markdown [公式ページ](https://huggingface.co/teasan/WeddingImperial)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/rBGYRNmznk8)
use_wedding_imperial_v2 = False #@param {type: "boolean"}
if use_wedding_imperial_v2:
  !wget https://huggingface.co/teasan/WeddingImperial/resolve/main/model/weddingImperialV2.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### Photon
#@markdown [公式ページ](https://civitai.com/models/84728/photon)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/eIYkdfiGx-Y)
use_photon = False #@param {type: "boolean"}
if use_photon:
  !wget https://civitai.com/api/download/models/90072?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### DreamShaper v7
#@markdown [公式ページ](https://huggingface.co/Lykon/DreamShaper)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/u071zv1lxug)
use_dream_shaper_v7 = False #@param {type: "boolean"}
if use_dream_shaper_v7:
  !wget https://huggingface.co/Lykon/DreamShaper/resolve/main/DreamShaper_7_pruned.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### DreamShaper v8
#@markdown [公式ページ](https://huggingface.co/Lykon/DreamShaper)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/gIfmr12C4X8)
use_dream_shaper_v8 = False #@param {type: "boolean"}
if use_dream_shaper_v8:
  !wget https://huggingface.co/Lykon/DreamShaper/resolve/main/DreamShaper_8_pruned.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### AbsoluteReality v1.6
#@markdown [公式ページ](https://huggingface.co/Lykon/AbsoluteReality)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/LlMlnkorC_k)
use_absolute_reality_v16 = False #@param {type: "boolean"}
if use_absolute_reality_v16:
  !wget https://huggingface.co/Lykon/AbsoluteReality/resolve/main/AbsoluteRealityV1.6525.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### AbsoluteReality v1.8.1
#@markdown [公式ページ](https://huggingface.co/Lykon/AbsoluteReality)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/vUr99l5UReA)
use_absolute_reality_v181 = False #@param {type: "boolean"}
if use_absolute_reality_v181:
  !wget https://huggingface.co/Lykon/AbsoluteReality/resolve/main/AbsoluteReality_1.8.1_pruned.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### Soda Mix
#@markdown [公式ページ](https://civitai.com/models/47507/soda-mix)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/2QGdRREYMoo)
use_soda_mix = False #@param {type: "boolean"}
if use_soda_mix:
  !wget https://civitai.com/api/download/models/52797?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### real-max-v3.4（补档）
#@markdown [公式ページ](https://civitai.com/models/60188/real-max-v34)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/PY8o7U4laSs)
use_real_max_v34 = False #@param {type: "boolean"}
if use_real_max_v34:
  !wget https://civitai.com/api/download/models/64660?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### BRAV7
#@markdown [公式ページ](https://civitai.com/models/25494?modelVersionId=177164)
use_bra_v7 = False #@param {type: "boolean"}
if use_bra_v7:
  !wget https://civitai.com/api/download/models/177164?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### BRAV6
#@markdown [公式ページ](https://huggingface.co/BanKaiPls/AsianModel)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/vcQplWWdyEk)
use_bra_v6 = False #@param {type: "boolean"}
if use_bra_v6:
  !wget https://huggingface.co/BanKaiPls/AsianModel/resolve/main/Brav6.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### BRAV5
#@markdown [公式ページ](https://huggingface.co/BanKaiPls/AsianModel)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/g0i1_LSzrWQ)
use_bra_v5 = False #@param {type: "boolean"}
if use_bra_v5:
  !wget https://huggingface.co/BanKaiPls/AsianModel/resolve/main/BRAV5finalfp16.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### BRAV4
#@markdown [公式ページ](https://civitai.com/models/25494?modelVersionId=51395)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/t6Nm0AUyqyU)
use_bra_v4 = False #@param {type: "boolean"}
if use_bra_v4:
  !wget https://civitai.com/api/download/models/51395?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### OpenBra
#@markdown [公式ページ](https://huggingface.co/BanKaiPls/AsianModel)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/0ZK7O5EFqWA?si=nU1tMkXFv-cW6ZMA)
use_open_bra = False #@param {type: "boolean"}
if use_open_bra:
  !wget https://huggingface.co/BanKaiPls/AsianModel/resolve/main/OpenBra.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### MUSE_v1
#@markdown [公式ページ](https://civitai.com/models/13564/musev1)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/Rc6v_2rOICA)
use_muse_v1 = False #@param {type: "boolean"}
if use_muse_v1:
  !wget https://civitai.com/api/download/models/15980?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### majicMIX realistic(v2.5)
#@markdown [公式ページ](https://civitai.com/models/43331/majicmix-realistic)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/A9Gx28_r1RE)
use_majicmix_realistic_v6 = False #@param {type: "boolean"}
if use_majicmix_realistic_v6:
  !wget https://civitai.com/api/download/models/126470?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### bellyberry1
#@markdown [公式ページ](https://civitai.com/models/109458/bellyberry1)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/lNLn8gnR7SU?si=O243teqpETa8xnHp)
use_bellyberry1 = False #@param {type: "boolean"}
if use_bellyberry1:
  !wget https://civitai.com/api/download/models/139410?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### SweetMuse
#@markdown [公式ページ](https://civitai.com/models/81668/sweetmuse)
use_sweet_muse = False #@param {type: "boolean"}
if use_sweet_muse:
  !wget https://civitai.com/api/download/models/89839?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### RealBeautyMix
#@markdown [公式ページ](https://civitai.com/models/85156)
use_real_beauty_mix = False #@param {type: "boolean"}
if use_real_beauty_mix:
  !wget https://civitai.com/api/download/models/90505?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### epiCRealism
#@markdown [公式ページ](https://civitai.com/models/25694?modelVersionId=134065)
use_epi_c_realism = False #@param {type: "boolean"}
if use_epi_c_realism:
  !wget https://civitai.com/api/download/models/134065?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### Noble Mix Fix
#@markdown [公式ページ](https://civitai.com/models/65480/noble-mix-fix)
use_noble_mix_fix = False #@param {type: "boolean"}
if use_noble_mix_fix:
  !wget https://civitai.com/api/download/models/92828?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### BlazingRealDrive
#@markdown [公式ページ](https://civitai.com/models/137909)
use_blazing_real_drive = False #@param {type: "boolean"}
if use_blazing_real_drive:
  !wget https://civitai.com/api/download/models/202466?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### VoidnoiseCore(R1311)
#@markdown [公式ページ](https://huggingface.co/konapieces/VoidnoiseCore)
use_voidnoise_core_r1311 = False #@param {type: "boolean"}
if use_voidnoise_core_r1311:
  !wget https://huggingface.co/konapieces/VoidnoiseCore/resolve/main/models/R1311/VoidnoiseCore_R1311.fp16.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### Kawaii Realistic Asian Mix
#@markdown [公式ページ](https://civitai.com/models/83766/kawaii-realistic-asian-mix)
use_kawaii_realistic_asian_mix = False #@param {type: "boolean"}
if use_kawaii_realistic_asian_mix:
  !wget https://civitai.com/api/download/models/170762?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### Meichidark_Mix
#@markdown [公式ページ](https://civitai.com/models/69158/meichidarkmix)
use_meichidark_mix = False #@param {type: "boolean"}
if use_meichidark_mix:
  !wget https://civitai.com/api/download/models/147184?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### CyberRealistic
#@markdown [公式ページ](https://civitai.com/models/15003/cyberrealistic)
use_cyber_realistic = False #@param {type: "boolean"}
if use_cyber_realistic:
  !wget https://civitai.com/api/download/models/198401?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/
  # CyberRealistic Negativeも併せてダウンロード
  !wget https://civitai.com/api/download/models/82745?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/embeddings/

#@markdown ### Realistic Vision
#@markdown [公式ページ](https://civitai.com/models/4201)
use_realistic_vision = False #@param {type: "boolean"}
if use_realistic_vision:
  !wget https://civitai.com/api/download/models/130072?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### XXMix_9realistic
#@markdown [公式ページ](https://civitai.com/models/47274/xxmix9realistic)
use_xxmix_9realistic = False #@param {type: "boolean"}
if use_xxmix_9realistic:
  !wget https://civitai.com/api/download/models/102222?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### BellyUse
#@markdown [公式ページ](https://civitai.com/models/211708)
use_belly_use = False #@param {type: "boolean"}
if use_belly_use:
  !wget https://civitai.com/api/download/models/265163?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### BlueberryMix
#@markdown [公式ページ](https://civitai.com/models/14323/blueberrymix)
use_blueberrymix = False #@param {type: "boolean"}
if use_blueberrymix:
  !wget https://civitai.com/api/download/models/16859?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### ICBINP
#@markdown [公式ページ](https://civitai.com/models/28059/icbinp-i-cant-believe-its-not-photography)
use_icbinp = False #@param {type: "boolean"}
if use_icbinp:
  !wget https://civitai.com/api/download/models/121557?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### ICBINP - LCM
#@markdown [公式ページ](https://civitai.com/models/28059/icbinp-i-cant-believe-its-not-photography)
use_icbinp_lcm = False #@param {type: "boolean"}
if use_icbinp_lcm:
  !wget https://civitai.com/api/download/models/253668?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### t3
#@markdown [公式ページ](https://civitai.com/models/110053/t3)
use_t3 = False #@param {type: "boolean"}
if use_t3:
  !wget https://civitai.com/api/download/models/281273?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ---
#@markdown ## 実写系モデル (Base: SD 2.1)

#@markdown ### Manga Diffusion PoC
#@markdown [公式ページ](https://huggingface.co/aipicasso/manga-diffusion-poc)
use_manga_diffusion_poc = False #@param {type: "boolean"}
if use_manga_diffusion_poc:
  !wget https://huggingface.co/aipicasso/manga-diffusion-poc/resolve/main/poc.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/


#@markdown ---
#@markdown ## 実写系モデル (Base: SDXL)

#@markdown ### SDXL base 1.0
#@markdown [公式ページ](https://huggingface.co/stabilityai/stable-diffusion-xl-base-1.0)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/IyXvmlhjjgM)
use_sdxl_base_1 = False #@param {type: "boolean"}
if use_sdxl_base_1:
  !wget https://huggingface.co/stabilityai/stable-diffusion-xl-base-1.0/resolve/main/sd_xl_base_1.0.safetensors --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### SDXL 1.0 ArienMixXL Asian portrait 亚洲人像
#@markdown [公式ページ](https://civitai.com/models/118913/sdxl-10-arienmixxl-asian-portrait)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/UAUhscKUvp8)
use_trainasian = False #@param {type: "boolean"}
if use_trainasian:
  !wget https://civitai.com/api/download/models/448627?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/


#@markdown ### DreamShaper XL1.0
#@markdown [公式ページ](https://civitai.com/models/112902)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/UAUhscKUvp8)
use_dream_shaper_xl_1 = False #@param {type: "boolean"}
if use_dream_shaper_xl_1:
  !wget https://civitai.com/api/download/models/126688?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### iNiverse Mix(SFW & NSFW)
#@markdown [公式ページ](https://civitai.com/models/226533/iniverse-mixsfw-and-nsfw?modelVersionId=1150354)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/O2GNFwB2Ncg)
use_iniverse_mix_v1 = False #@param {type: "boolean"}
if use_iniverse_mix_v1:
  !wget https://civitai.com/api/download/models/1150354?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### haduki_mix
#@markdown [公式ページ](https://huggingface.co/Kotajiro/haduki_mix)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/LrSyTqXa_bU?si=PCKaBIJSeG7uGxcB)
use_haduki_mix = False #@param {type: "boolean"}
if use_haduki_mix:
  !wget https://huggingface.co/Kotajiro/haduki_mix/resolve/main/haduki_mix_v10_FP16.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### nagatsuki_mix
#@markdown [公式ページ](https://huggingface.co/Kotajiro/nagatsuki_mix)
use_nagatsuki_mix = False #@param {type: "boolean"}
if use_nagatsuki_mix:
   !wget https://civitai.com/api/download/models/238354?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### Copax TimeLessXL - SDXL1.0
#@markdown [公式ページ](https://civitai.com/models/118111/copax-timelessxl-sdxl10)
use_copax_timelessxl_sdxl10 = False #@param {type: "boolean"}
if use_copax_timelessxl_sdxl10:
  !wget https://civitai.com/api/download/models/172160?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### AfterRealXL
#@markdown [公式ページ](https://civitai.com/models/150212/afterrealxl)
use_after_real_xl = False #@param {type: "boolean"}
if use_after_real_xl:
  !wget https://civitai.com/api/download/models/167881?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### NightVision XL
#@markdown [公式ページ](https://civitai.com/models/128607/nightvision-xl-photorealistic-or-portrait-or-photography-or-hyperreal-or-architecture-or-interior-design-no-refiner-needed)
use_nightvision_xl = False #@param {type: "boolean"}
if use_nightvision_xl:
  !wget https://civitai.com/api/download/models/159982?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### Latent Consistency Model (LCM): SSD-1B
#@markdown [公式ページ](https://huggingface.co/latent-consistency/lcm-ssd-1b)
use_lcm_ssd_1b = False #@param {type: "boolean"}
if use_lcm_ssd_1b:
  !wget https://huggingface.co/latent-consistency/lcm-ssd-1b/resolve/main/diffusion_pytorch_model.fp16.safetensors --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### Imaginarium (Turbo boosted)
#@markdown [公式ページ](https://civitai.com/models/249973/imaginarium?modelVersionId=316617)
use_imaginarium_turbo_boosted = False #@param {type: "boolean"}
if use_imaginarium_turbo_boosted:
  !wget https://civitai.com/api/download/models/316617?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### SDXL-Lightning (1step)
#@markdown [公式ページ](https://huggingface.co/ByteDance/SDXL-Lightning)
use_sdxl_lightning_1step = False #@param {type: "boolean"}
if use_sdxl_lightning_1step:
  !wget https://huggingface.co/ByteDance/SDXL-Lightning/resolve/main/sdxl_lightning_1step_x0.safetensors --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### SDXL-Lightning (2step)
#@markdown [公式ページ](https://huggingface.co/ByteDance/SDXL-Lightning)
use_sdxl_lightning_2step = False #@param {type: "boolean"}
if use_sdxl_lightning_2step:
  !wget https://huggingface.co/ByteDance/SDXL-Lightning/resolve/main/sdxl_lightning_2step.safetensors --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### SDXL-Lightning (4step)
#@markdown [公式ページ](https://huggingface.co/ByteDance/SDXL-Lightning)
use_sdxl_lightning_4step = False #@param {type: "boolean"}
if use_sdxl_lightning_4step:
  !wget https://huggingface.co/ByteDance/SDXL-Lightning/resolve/main/sdxl_lightning_4step.safetensors --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### SDXL-Lightning (8step)
#@markdown [公式ページ](https://huggingface.co/ByteDance/SDXL-Lightning)
use_sdxl_lightning_8step = False #@param {type: "boolean"}
if use_sdxl_lightning_8step:
  !wget https://huggingface.co/ByteDance/SDXL-Lightning/resolve/main/sdxl_lightning_8step.safetensors --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### DreamShaper XL (Lightning DPM++ SDE)
#@markdown [公式ページ](https://civitai.com/models/112902?modelVersionId=354657)
use_dream_shaper_xl_lightning_dpm_sde = False #@param {type: "boolean"}
if use_dream_shaper_xl_lightning_dpm_sde:
  !wget https://civitai.com/api/download/models/354657?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### Juggernaut XL (V9+RDPhoto2-Lightning_4S)
#@markdown [公式ページ](https://civitai.com/models/133005/juggernaut-xl)
use_juggernaut_xl_v9_rdphoto2_lightning_4s = False #@param {type: "boolean"}
if use_juggernaut_xl_v9_rdphoto2_lightning_4s:
  !wget https://civitai.com/api/download/models/357609?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### Pony Realism
#@markdown [公式ページ](https://civitai.com/models/372465?modelVersionId=914390)
use_PonyRealism = False #@param {type: "boolean"}
if use_PonyRealism:
  !wget https://civitai.com/api/download/models/914390?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/


#@markdown ### PinkiePie pony mix
#@markdown [公式ページ](https://civitai.com/models/457669?modelVersionId=586928)
use_PinkiePieponymix = False #@param {type: "boolean"}
if use_PinkiePieponymix:
  !wget https://civitai.com/api/download/models/586928?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### Asian_Realism_By_Stable_Yogi (PONY)
#@markdown [公式ページ](https://civitai.com/models/168949/asianrealismbystableyogi-pony)
use_Realismpony = False #@param {type: "boolean"}
if use_Realismpony:
  !wget https://civitai.com/api/download/models/1624625?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### RealMixPony
#@markdown [公式ページ](https://civitai.com/models/489668/realmixpony)
use_RealMixPony = False #@param {type: "boolean"}
if use_RealMixPony:
  !wget https://civitai.com/api/download/models/1657892?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### YaminabePony
#@markdown [公式ページ](https://civitai.com/models/409856?modelVersionId=735173)
use_YaminabePony = False #@param {type: "boolean"}
if use_YaminabePony:
  !wget https://civitai.com/api/download/models/735173?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/


#@markdown ### Kawaii Realistic Asian Pony Mix + UwazumiMix
#@markdown [公式ページ](https://civitai.com/models/1463974/kawaii-realistic-asian-pony-mix-uwazumimix)
use_Kawaii = False #@param {type: "boolean"}
if use_Kawaii:
  !wget https://civitai.com/api/download/models/1655688?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/


#@markdown ### iniverse
#@markdown [公式ページ](https://civitai.com/models/1463974/kawaii-realistic-asian-pony-mix-uwazumimix)
use_iniverse = False #@param {type: "boolean"}
if use_iniverse:
  !wget https://civitai.com/api/download/models/1366495?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/


#@markdown ### RunBullXL - Pony based Photographic Model
#@markdown [公式ページ](https://civitai.com/models/348487/runbullxl-pony-based-photographic-model)
use_runbull = False #@param {type: "boolean"}
if use_runbull:
  !wget https://civitai.com/api/download/models/1538985?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### WAI-REALMIX
#@markdown [公式ページ](https://civitai.com/models/393905/wai-realmix)
use_waireal = False #@param {type: "boolean"}
if use_waireal:
  !wget https://civitai.com/api/download/models/868204?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### HARUKI_MIX
#@markdown [公式ページ](https://civitai.com/models/856375/harukimix)
use_HARUKI = False #@param {type: "boolean"}
if use_HARUKI:
  !wget https://civitai.com/api/download/models/2099686?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### CyberRealistic Pony
#@markdown [公式ページ](https://civitai.com/models/443821?modelVersionId=1177183)
use_CyberRealistic = False #@param {type: "boolean"}
if use_CyberRealistic:
  !wget https://civitai.com/api/download/models/1177183?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### Fucktastic Real Checkpoint
#@markdown [公式ページ](https://civitai.red/models/918104/fucktastic-real-checkpoint-ponypdxl-porn-realistic-nsfwsfw?modelVersionId=2320801)
use_FucktasticReal = False #@param {type: "boolean"}
if use_FucktasticReal:
  !wget https://civitai.com/api/download/models/2320801?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/






#@markdown ---
#@markdown ## イラスト系モデル (Base: SD 1.5)

#@markdown ### hassaku_hentai_model
#@markdown [公式ページ](https://civitai.com/models/2583/hassaku-hentai-model)

#@markdown [※R指定モデル]
use_hassaku_hentai_model = False #@param {type: "boolean"}
if use_hassaku_hentai_model:
  !wget https://civitai.com/api/download/models/106922?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### umaaji
#@markdown [公式ページ](https://civitai.com/models/42912?modelVersionId=165855)

#@markdown [※R指定モデル]
use_umaaji = False #@param {type: "boolean"}
if use_umaaji:
  !wget https://civitai.com/api/download/models/165855?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/


#@markdown ### Dark_Sushi_Mix_大颗寿司Mix
#@markdown [公式ページ](https://civitai.com/models/24779/dark-sushi-mix-mix)

#@markdown [※]
use_Dark_Sushi_Mix_大颗寿司Mix = False #@param {type: "boolean"}
if use_Dark_Sushi_Mix_大颗寿司Mix:
  !wget https://civitai.com/api/download/models/93208?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/


#@markdown ### seizaMix
#@markdown [公式ページ](https://civitai.com/models/116279?modelVersionId=125903)

#@markdown [※]
use_seizaMix = False #@param {type: "boolean"}
if use_seizaMix:
  !wget https://civitai.com/api/download/models/125903?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### Sweet_mix
#@markdown [公式ページ](https://civitai.com/models/18927/sweet-mix)

#@markdown [※]
use_Sweet_mix = False #@param {type: "boolean"}
if use_Sweet_mix:
  !wget https://civitai.com/api/download/models/158294?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### flatpiececore
#@markdown [公式ページ](https://civitai.com/models/157686/flatpiececore?modelVersionId=264383)

#@markdown [※]
use_flatpiececore = False #@param {type: "boolean"}
if use_flatpiececore:
  !wget https://civitai.com/api/download/models/264383?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### MomijiMix
#@markdown [公式ページ](https://civitai.com/models/176410/momijimix)

#@markdown [※]
use_MomijiMix = False #@param {type: "boolean"}
if use_MomijiMix:
  !wget https://civitai.com/api/download/models/198066?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### meinamix
#@markdown [公式ページ](https://civitai.com/models/7240/meinamix)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/LX7_3vmUImY)
use_meina_mix = False #@param {type: "boolean"}
if use_meina_mix:
  !wget https://civitai.com/api/download/models/119057?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### DMCMIX
#@markdown [公式ページ](https://civitai.com/models/250036/dmc-mix)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/LX7_3vmUImY)
use_DMCMIX = False #@param {type: "boolean"}
if use_DMCMIX:
  !wget https://civitai.com/api/download/models/282159?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### Tsubaki
#@markdown [公式ページ](https://civitai.com/models/58557/tsubaki)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/LX7_3vmUImY)
use_Tsubaki = False #@param {type: "boolean"}
if use_Tsubaki:
  !wget https://civitai.com/api/download/models/173636?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### Anything
#@markdown [公式ページ](https://civitai.com/models/9409?modelVersionId=30163)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/LX7_3vmUImY)
use_Anything = False #@param {type: "boolean"}
if use_Anything:
  !wget https://civitai.com/api/download/models/30163?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### YesMix
#@markdown [公式ページ](https://civitai.com/models/9139?modelVersionId=30163)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/LX7_3vmUImY)
use_YesMix = False #@param {type: "boolean"}
if use_YesMix:
  !wget https://civitai.com/api/download/models/11086?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### aom3
#@markdown [公式ページ](https://civitai.com/models/9942/abyssorangemix3-aom3)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/LX7_3vmUImY)
use_aom3 = False #@param {type: "boolean"}
if use_aom3:
  !wget https://civitai.com/api/download/models/17233?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/


#@markdown ### Anime_Pastel_Dream
#@markdown [公式ページ](https://civitai.com/models/196818/anime-dreamer-mixed)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/LX7_3vmUImY)
use_Anime_Pastel_Dream = False #@param {type: "boolean"}
if use_Anime_Pastel_Dream:
  !wget https://civitai.com/api/download/models/28100?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/


#@markdown ### BraveDawn
#@markdown [公式ページ](https://civitai.com/models/101932)

#@markdown VAEの `vae-ft-mse-840000-ema-pruned.ckpt` を併せて利用することが公式より推奨されています
use_brave_dawn = False #@param {type: "boolean"}
if use_brave_dawn:
  !wget https://civitai.com/api/download/models/128200?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### Break the Darkness
#@markdown [公式ページ](https://civitai.com/models/105047/break-the-darkness)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/Slpt9bQ6884)
use_break_the_darkness = False #@param {type: "boolean"}
if use_break_the_darkness:
  !wget https://civitai.com/api/download/models/123633?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### FlexDreamHK
#@markdown [公式ページ](https://huggingface.co/den2nova/FlexDreamHK)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/wiG9GbXpPis)
use_flex_dream_hk = False #@param {type: "boolean"}
if use_flex_dream_hk:
  !wget https://huggingface.co/den2nova/FlexDreamHK/resolve/main/FlexDreamHK_v2.0.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### ENCartoony
#@markdown [公式ページ](https://civitai.com/models/143113)

use_encartoony = False #@param {type: "boolean"}
if use_encartoony:
  !wget https://civitai.com/api/download/models/158785?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### ENCLight-v1
#@markdown [公式ページ](https://civitai.com/models/161452/enclight-v1)

use_enclight_v1 = False #@param {type: "boolean"}
if use_enclight_v1:
  !wget https://civitai.com/api/download/models/181723?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### Flat-2D Animerge
#@markdown [公式ページ](https://civitai.com/models/35960/flat-2d-animerge)

use_flat2d_animerge = False #@param {type: "boolean"}
if use_flat2d_animerge:
  !wget https://civitai.com/api/download/models/266360?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### AnyLoRA - Checkpoint
#@markdown [公式ページ](https://civitai.com/models/23900/anylora-checkpoint)

use_anylora_checkpoint = False #@param {type: "boolean"}
if use_anylora_checkpoint:
  !wget https://civitai.com/api/download/models/95489?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### Mistoon_Anime
#@markdown [公式ページ](https://civitai.com/models/24149/mistoonanime?modelVersionId=348981)

use_Mistoon_checkpoint = False #@param {type: "boolean"}
if use_Mistoon_checkpoint:
  !wget https://civitai.com/api/download/models/348981?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/


#@markdown ---
#@markdown ## イラスト系モデル (Base: SDXL)

#@markdown ### BreakDomainXL
#@markdown [公式ページ](https://civitai.com/models/126259/breakdomainxl)

#@markdown VAEの `sdxl_vae` を併せて利用することが公式より推奨されています

use_break_domain_xl = False #@param {type: "boolean"}
if use_break_domain_xl:
  !wget https://civitai.com/api/download/models/145893?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### LastpieceCoreXL
#@markdown [公式ページ](https://huggingface.co/konapieces/LastpieceCoreXL)
use_last_piece_core_xl = False #@param {type: "boolean"}
if use_last_piece_core_xl:
  !wget https://huggingface.co/konapieces/LastpieceCoreXL/resolve/main/models/S0964/LastpieceCoreXL_S0964.fp16.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/
  # 専用のVAEも併せて導入
  !wget https://huggingface.co/konapieces/LastpieceCoreXL/resolve/main/vae/lastpiece.xl.vae.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/VAE/

#@markdown ### blue_pencil-XL
#@markdown [公式ページ](https://civitai.com/models/119012/bluepencil-xl)
use_blue_pencil_xl = False #@param {type: "boolean"}
if use_blue_pencil_xl:
  !wget https://civitai.com/api/download/models/212090?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/
  # 推奨されるNegative Embeddingsも導入
  # unaestheticXL | Negative TI (_Jug6, _Sky3.1, _AYv1)
  # URL: https://civitai.com/models/119032/unaestheticxl-or-negative-ti)
  !wget https://civitai.com/api/download/models/207934?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/embeddings/
  !wget https://civitai.com/api/download/models/175819?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/embeddings/
  !wget https://civitai.com/api/download/models/162146?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/embeddings/

#@markdown ### blue_pencil-XL-LCM
#@markdown [公式ページ](https://civitai.com/models/202108/bluepencil-xl-lcm)
use_blue_pencil_xl_lcm = False #@param {type: "boolean"}
if use_blue_pencil_xl_lcm:
  !wget https://civitai.com/api/download/models/227693?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### NekoRayXL
#@markdown [公式ページ](https://civitai.com/models/136719/nekorayxl)
use_neko_ray_xl = False #@param {type: "boolean"}
if use_neko_ray_xl:
  !wget https://civitai.com/api/download/models/150826?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### Animagine XL V3
#@markdown [公式ページ](https://civitai.com/models/260267/animagine-xl-v3)
use_animagine_xl_v3 = False #@param {type: "boolean"}
if use_animagine_xl_v3:
  !wget https://civitai.com/api/download/models/293564?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### WAI-NSFW-illustrious-SDXLv8.0
#@markdown [公式ページ](https://civitai.com/models/827184?modelVersionId=1183765)
use_wainsfw = False #@param {type: "boolean"}
if use_wainsfw:
  !wget https://civitai.com/api/download/models/1183765?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### WAI-NSFW-illustrious-SDXLv17.0
#@markdown [公式ページ](https://civitai.red/models/827184/wai-illustrious-sdxl?modelVersionId=2883731)
use_wainsfw2 = True #@param {type: "boolean"}
if use_wainsfw2:
  !wget https://civitai.com/api/download/models/2883731?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### WAI-illustrious-SDXL
#@markdown [公式ページ](https://civitai.com/models/827184?modelVersionId=1761560)
use_waiillust = False #@param {type: "boolean"}
if use_waiillust:
  !wget https://civitai.com/api/download/models/2514310?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### illustrious-XL
#@markdown [公式ページ](https://civitai.com/models/795765?modelVersionId=889818)
use_illust = False #@param {type: "boolean"}
if use_illust:
  !wget https://civitai.com/api/download/models/889818?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/


#@markdown ### AutismMix SDXL
#@markdown [公式ページ](https://civitai.com/models/288584?modelVersionId=324619)
use_AutismMix = False #@param {type: "boolean"}
if use_AutismMix:
  !wget https://civitai.com/api/download/models/324619?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

#@markdown ### Pony Diffusion V6 XL
#@markdown [公式ページ](https://civitai.com/models/257749/pony-diffusion-v6-xl)
use_Diffusion = False #@param {type: "boolean"}
if use_Diffusion:
  !wget https://civitai.com/api/download/models/290640?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/



#@markdown ### Hassaku XL (Pony)
#@markdown [公式ページ](https://civitai.com/models/376031/hassaku-xl-pony)
use_hassaku = False #@param {type: "boolean"}
if use_hassaku:
  !wget https://civitai.com/api/download/models/575495?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/


#@markdown ### Illustrious XL personal merge (noob v-pred0.5 test merge updated)
#@markdown [公式ページ](https://civitai.com/models/835655?modelVersionId=1023901)
use_Illustrious = False #@param {type: "boolean"}
if use_Illustrious:
  !wget https://civitai.com/api/download/models/1023901?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/







#@markdown ---
#@markdown ## セミリアル系モデル (Base: SD 1.5)

#@markdown ### LittleStepMix v1
#@markdown LittleStepMixを導入する際は併せて、【sr_SDv2vae_kl-f8anime2.safetensors】というVAEも一緒にインストールします

#@markdown [公式ページ](https://huggingface.co/sazyou-roukaku/LittleStepMix)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/a2X3oJB55nE)
use_little_step_mix_v1 = False #@param {type: "boolean"}
if use_little_step_mix_v1:
  # モデルファイルのダウンロード(LittleStepMix_v1)
  !wget https://huggingface.co/sazyou-roukaku/LittleStepMix/resolve/main/LittleStepMix_v1.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/

  # VAEのダウンロード(sr_SDv2vae_kl-f8anime2.safetensors) - LittleStepMix_v1に必要
  !wget https://huggingface.co/sazyou-roukaku/LittleStepMix/resolve/main/VAE/sr_SDv2vae_kl-f8anime2.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/VAE/


#@markdown ---
#@markdown ## Stable Video Diffusion

#@markdown ### StableVideoDiffusion（fp16版）
#@markdown [公式ページ](https://huggingface.co/becausecurious/stable-video-diffusion-img2vid-fp16)

use_stable_video_diffusion_fp16 = False #@param {type: "boolean"}
if use_stable_video_diffusion_fp16:
  !wget https://huggingface.co/becausecurious/stable-video-diffusion-img2vid-fp16/resolve/main/svd_xt-fp16.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/svd/


#@markdown ---
#@markdown ## refiner向けモデル (SDXL向け)

#@markdown ### SD-XL 1.0-refiner
#@markdown [公式ページ](https://huggingface.co/stabilityai/stable-diffusion-xl-refiner-1.0)

#@markdown [生成される画像を確認する(YouTube)](https://youtu.be/CNhrbc8VRSk?si=OSQVldOp7FMU--zL)
use_sdxl_1_refiner = False #@param {type: "boolean"}
if use_sdxl_1_refiner:
  !wget https://huggingface.co/stabilityai/stable-diffusion-xl-refiner-1.0/resolve/main/sd_xl_refiner_1.0.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/Stable-diffusion/



#@markdown ---
#@markdown ## 依頼関連

#@markdown ### Aesthetic Quality Modifiers - Masterpiece
#@markdown [公式ページ](https://civitai.com/models/929497?modelVersionId=1050644)
use_Masterpiece = True #@param {type: "boolean"}
if use_Masterpiece:
  !wget https://civitai.com/api/download/models/1050644?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Aesthetic Quality Modifiers - Best Quality
#@markdown [公式ページ](https://civitai.com/models/977115?modelVersionId=1116960)
use_Quality = True #@param {type: "boolean"}
if use_Quality:
  !wget https://civitai.com/api/download/models/1116960?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/


#@markdown ### AIイラストおじさん style illustrious | goofy Ai
#@markdown [公式ページ](https://civitai.com/models/1001550/ai-style-illustrious-or-goofy-ai)
use_ojisan2 = True #@param {type: "boolean"}
if use_ojisan2:
  !wget https://civitai.com/api/download/models/1122510?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/


#@markdown ### MItsuha Nakano (中野 三玖) - The Quintessential Quintuplets (Gotoubun no Hanayome) (五等分の花嫁)
#@markdown [公式ページ](https://civitai.com/models/1337567?modelVersionId=1510447)
use_Mitsuha = False #@param {type: "boolean"}
if use_Mitsuha:
  !wget https://civitai.com/api/download/models/1510447?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/


#@markdown ### Yotsuba Nakano (中野 四葉) - The Quintessential Quintuplets (Gotoubun no Hanayome) (五等分の花嫁)
#@markdown [公式ページ](https://civitai.com/models/1337568?modelVersionId=1510448)
use_Yotsuba = False #@param {type: "boolean"}
if use_Yotsuba:
  !wget https://civitai.com/api/download/models/1510448?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Itsuki Nakano (中野 五月)- The Quintessential Quintuplets (Gotoubun no Hanayome) (五等分の花嫁)
#@markdown [公式ページ](https://civitai.com/models/1337566?modelVersionId=1510445)
use_itsuki = False #@param {type: "boolean"}
if use_itsuki:
  !wget https://civitai.com/api/download/models/1510445?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Aqua (アクア) - KonoSuba: God's Blessing on This Wonderful World! (この素晴らしい世界に祝福を！)
#@markdown [公式ページ](https://civitai.com/models/1376640/aqua-konosuba-gods-blessing-on-this-wonderful-world)
use_Aqua = False #@param {type: "boolean"}
if use_Aqua:
  !wget https://civitai.com/api/download/models/1555410?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Megumin (めぐみん) - KonoSuba: God's Blessing on This Wonderful World! (この素晴らしい世界に祝福を！)
#@markdown [公式ページ](https://civitai.com/models/1376641/megumin-konosuba-gods-blessing-on-this-wonderful-world)
use_Megumin = False #@param {type: "boolean"}
if use_Megumin:
  !wget https://civitai.com/api/download/models/1555411?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/


#@markdown ### Yukinoshita Yukino [5 outfits] | Illustrious | Yahari Ore no Seishun Love Comedy wa Machigatteiru.
#@markdown [公式ページ](https://civitai.com/models/1302624/yukinoshita-yukino-5-outfits-or-illustrious-or-yahari-ore-no-seishun-love-comedy-wa-machigatteiru)
use_yukinoshita = False #@param {type: "boolean"}
if use_yukinoshita:
  !wget https://civitai.com/api/download/models/1470350?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/


#@markdown ### Yuigahama Yui [5 outfits] | Illustrious | Yahari Ore no Seishun Love Comedy wa Machigatteiru.
#@markdown [公式ページ](https://civitai.com/models/1302638/yuigahama-yui-5-outfits-or-illustrious-or-yahari-ore-no-seishun-love-comedy-wa-machigatteiru)
use_Yuigahama = False #@param {type: "boolean"}
if use_Yuigahama:
  !wget https://civitai.com/api/download/models/1470367?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Akane Hououji (鳳凰寺 紅葉) - The Café Terrace and Its Goddesses (Megami no Café Terrace) (女神のカフェテラス)
#@markdown [公式ページ](https://civitai.com/models/1964153/akane-hououji-the-cafe-terrace-and-its-goddesses-megami-no-cafe-terrace-commission)
use_Akane = False #@param {type: "boolean"}
if use_Akane:
  !wget https://civitai.com/api/download/models/2223209?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Shiragiku "Kiku-chan" Ono (小野 白菊) - The Café Terrace and Its Goddesses (Megami no Café Terrace) (女神のカフェテラス)
#@markdown [公式ページ]()
use_Shiragiku = False #@param {type: "boolean"}
if use_Shiragiku:
  !wget https://civitai.com/api/download/models/2223348?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Kuze Masachika / Roshidere
#@markdown [公式ページ](https://civitai.com/models/1729333/kuze-masachika-roshidere)
use_Masachika = False #@param {type: "boolean"}
if use_Masachika:
  !wget https://civitai.com/api/download/models/1957148?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Yuki Suou | Alya Sometimes Hides Her Feelings in Russian | 時々ボソッとロシア語でデレる隣のアーリャさん
#@markdown [公式ページ](https://civitai.com/models/1000015)
use_Suou = False #@param {type: "boolean"}
if use_Suou:
  !wget https://civitai.com/api/download/models/1120706?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Alisa Mikhailovna Kujou (Ayra-san) | Alya Sometimes Hides Her Feelings in Russian | 時々ボソッとロシア語でデレる隣のアーリャ
#@markdown [公式ページ](https://civitai.com/models/955584/alisa-mikhailovna-kujou-ayra-san-or-alya-sometimes-hides-her-feelings-in-russian-or)
use_Alisa = False #@param {type: "boolean"}
if use_Alisa:
  !wget https://civitai.com/api/download/models/1069879?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Maria Mikhailovna Kujou (Masha) | Alya Sometimes Hides Her Feelings in Russian | 時々ボソッとロシア語でデレる隣のアーリャさん
#@markdown [公式ページ](https://civitai.com/models/971431)
use_Maria = False #@param {type: "boolean"}
if use_Maria:
  !wget https://civitai.com/api/download/models/1087752?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/


#@markdown ### Misaki Shokuhou (食蜂 操祈) - A Certain Scientific Railgun (とある科学の超電磁砲)
#@markdown [公式ページ](https://civitai.com/models/1354311/misaki-shokuhou-a-certain-scientific-railgun)
use_Shokuhou = False #@param {type: "boolean"}
if use_Shokuhou:
  !wget https://civitai.com/api/download/models/1529814?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/


#@markdown ### Mikoto Misaka (御坂 美琴) - A Certain Scientific Railgun (とある科学の超電磁砲)
#@markdown [公式ページ](https://civitai.com/models/1354309/mikoto-misaka-a-certain-scientific-railgun)
use_MikotoMisaka = False #@param {type: "boolean"}
if use_MikotoMisaka:
  !wget https://civitai.com/api/download/models/1529810?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### [ILLUST] Mizuhara Chizuru | Kanojo, Okarishimasu
#@markdown [公式ページ](https://civitai.com/models/1409014/illust-mizuhara-chizuru-or-kanojo-okarishimasu)
use_MizuharaChizuru = False #@param {type: "boolean"}
if use_MizuharaChizuru:
  !wget https://civitai.com/api/download/models/1592793?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### [ILLUST] Sumi Sakurasawa | Kanojo, Okarishimasu
#@markdown [公式ページ](https://civitai.com/models/1844961)
use_SumiSakurasawa = False #@param {type: "boolean"}
if use_SumiSakurasawa:
  !wget https://civitai.com/api/download/models/2087878?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### [ILLUST] Ruka Sarashina | Kanojo Okarishimasu
#@markdown [公式ページ](https://civitai.com/models/1192047)
use_RukaSarashina = False #@param {type: "boolean"}
if use_RukaSarashina:
  !wget https://civitai.com/api/download/models/1342106?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/



#@markdown ### Kyouan Style
#@markdown [公式ページ](https://civitai.com/models/815974?modelVersionId=1283809)
use_Kyouan = True #@param {type: "boolean"}
if use_Kyouan:
  !wget https://civitai.com/api/download/models/1283809?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Enchanting Eyes (Detailed Eyes)
#@markdown [公式ページ](https://civitai.com/models/974076/enchanting-eyes-detailed-eyes)
use_enchanting = True #@param {type: "boolean"}
if use_enchanting:
  !wget https://civitai.com/api/download/models/1463317?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Empty Eyes [Illustrious]
#@markdown [公式ページ](https://civitai.com/models/934416?modelVersionId=1058910)
use_Empty = True #@param {type: "boolean"}
if use_Empty:
  !wget https://civitai.com/api/download/models/1058910?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ---
#@markdown ## その他オプション


#@markdown ### (PonyXL) Peeking Out (Concept / Pose)
#@markdown [公式ページ](https://civitai.com/models/350604/ponyxl-peeking-out-concept-pose)
use_Peeking = False #@param {type: "boolean"}
if use_Peeking:
  !wget https://civitai.com/api/download/models/392159?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/


#@markdown ### Doggy Fingering (Pony)
#@markdown [公式ページ](https://civitai.com/models/434900/doggy-fingering-pony)
use_Fingering = False #@param {type: "boolean"}
if use_Fingering:
  !wget https://civitai.com/api/download/models/484414?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/
# DoggyFingersER, 2girls, 1boy, POV, ass, doggystyle

#@markdown ### [JOJO] by Araki hirohiko style
#@markdown [公式ページ](https://civitai.com/models/607276/jojo-by-araki-hirohiko-style)
use_jojo = False #@param {type: "boolean"}
if use_jojo:
  !wget https://civitai.com/api/download/models/1684645?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/
# DoggyFingersER, 2girls, 1boy, POV, ass, doggystyle





#@markdown ---
#@markdown ## 欲求関連関連

#@markdown ### To Love Ru Darkness Comic Style - pony
#@markdown [公式ページ](https://civitai.com/models/880827/to-love-ru-darkness-comic-style-pony-or-to-love)
use_toto = False #@param {type: "boolean"}
if use_toto:
  !wget https://civitai.com/api/download/models/986008?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/


#@markdown ### FF10
#@markdown [公式ページ](https://civitai.com/models/1535450/final-fantasy-x-group)
use_ff10 = False #@param {type: "boolean"}
if use_ff10:
  !wget https://civitai.com/api/download/models/1737337?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Seiken Densetsu - Mana Series Style [SDXL+Pony]

#@markdown [公式ページ](https://civitai.com/models/311263/seiken-densetsu-mana-series-style-sdxlpony)
use_seiken3 = False #@param {type: "boolean"}
if use_seiken3:
  !wget https://civitai.com/api/download/models/369766?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/


#@markdown ### Toaru Kagaku no Railgun Style (Pony)
#@markdown [公式ページ](https://civitai.com/models/536191/toaru-kagaku-no-railgun-style-pony)
use_toaru = False #@param {type: "boolean"}
if use_toaru:
  !wget https://civitai.com/api/download/models/596066?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/


#@markdown ### DQ8 jeccica (Pony)
#@markdown [公式ページ](https://civitai.com/models/362136/jessica-albert-dragon-quest)
use_jessica = False #@param {type: "boolean"}
if use_jessica:
  !wget https://civitai.com/api/download/models/404666?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### DQ4 Arena (Pony)
#@markdown [公式ページ](https://civitai.com/models/362119/alena-dragon-quest)
use_Arena = False #@param {type: "boolean"}
if use_Arena:
  !wget https://civitai.com/api/download/models/404651?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/


#@markdown ### Daily Lives of My Countryside Anna
#@markdown [公式ページ](https://civitai.com/models/437916/daily-lives-of-my-countryside-character-collection)
use_Anna = False #@param {type: "boolean"}
if use_Anna:
  !wget https://civitai.com/api/download/models/2754571?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/


#@markdown ### The Quintessential Quintuplets
#@markdown [公式ページ](https://civitai.com/models/440742/the-quintessential-quintuplets-ponyxl-all-nakano-sisters-ichika-nino-miku-yotsuba-itsuki)
use_nakano5 = False #@param {type: "boolean"}
if use_nakano5:
  !wget https://civitai.com/api/download/models/490944?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Code Geass Style LoRA
#@markdown [公式ページ](https://civitai.com/models/90883/code-geass-style-lora)
use_geass= False #@param {type: "boolean"}
if use_geass:
  !wget https://civitai.com/api/download/models/96848?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Liz - The Tower and the Grimoire (Clymenia)

#@markdown [公式ページ](https://civitai.com/models/1748042/liz-the-tower-and-the-grimoire-clymenia)
use_liz= False #@param {type: "boolean"}
if use_liz:
  !wget https://civitai.com/api/download/models/1978361?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Pronant symhpony style

#@markdown [公式ページ](https://civitai.com/models/2073014/pronant-symhpony-style)
use_pronant= False #@param {type: "boolean"}
if use_pronant:
  !wget https://civitai.com/api/download/models/2345689?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/




#@markdown ---
#@markdown ## big brother 関連

#@markdown ### Big Brother Game Characters - Pony
#@markdown [公式ページ](https://civitai.com/models/1252565/big-brother-game-characters-pony)
use_BigBrother = False #@param {type: "boolean"}
if use_BigBrother:
  !wget https://civitai.com/api/download/models/1412099?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Alice from Big Brother Game - Pony
#@markdown [公式ページ](https://civitai.com/models/1172751?modelVersionId=1319522)
use_Alice = False #@param {type: "boolean"}
if use_Alice:
  !wget https://civitai.com/api/download/models/1319522?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Lisa from Big Brother Game - Pony
#@markdown [公式ページ](https://civitai.com/models/1153634/lisa-from-big-brother-game-pony)
use_Lisa = False #@param {type: "boolean"}
if use_Lisa:
  !wget https://civitai.com/api/download/models/1297611?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Ann from Big Brother Game - Pony
#@markdown [公式ページ](https://civitai.com/models/1193979/ann-from-big-brother-game-pony)
use_Ann = False #@param {type: "boolean"}
if use_Ann:
  !wget https://civitai.com/api/download/models/1344324?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Kira from Big Brother Game - Pony
#@markdown [公式ページ](https://civitai.com/models/1206056/kira-from-big-brother-game-pony)
use_Kira = False #@param {type: "boolean"}
if use_Kira:
  !wget https://civitai.com/api/download/models/1358177?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/



#@markdown ---
#@markdown ## lilitales 関連

#@markdown ### Lunalie el Blanca - Lilitales
#@markdown [公式ページ](https://civitai.com/models/1171682/lunalie-el-blanca-lilitales?modelVersionId=1318285)
use_luna = False #@param {type: "boolean"}
if use_luna:
  !wget https://civitai.com/api/download/models/1318285?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Lunalia2 - Lilitales
#@markdown [公式ページ](https://civitai.com/models/322918?modelVersionId=362044)
use_luna2 = False #@param {type: "boolean"}
if use_luna2:
  !wget https://civitai.com/api/download/models/362044?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/


#@markdown ### Jill Grantz - Lilitales
#@markdown [公式ページ](https://civitai.com/models/1177347/jill-grantz-lilitales)
use_Jill = False #@param {type: "boolean"}
if use_Jill:
  !wget https://civitai.com/api/download/models/1324795?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/


#@markdown ### Cless Gretchen クレス＝グレッチェン / Lilitales Lilipalace(kouyu 73gou Koudou)
#@markdown [公式ページ](https://civitai.com/models/1232102/cless-gretchen-lilitales)
use_Cless = False #@param {type: "boolean"}
if use_Cless:
  !wget https://civitai.com/api/download/models/1388351?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/


#@markdown ### Airi (Lilitales) LoRA
#@markdown [公式ページ](https://civitai.com/models/148309/airi-lilitales-lora)
use_Airi = False #@param {type: "boolean"}
if use_Airi:
  !wget https://civitai.com/api/download/models/165485?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/


#@markdown ### Anus UHD - pony
#@markdown [公式ページ](https://civitai.com/models/1201047/female-anus-uhd-pony)
use_anus = False #@param {type: "boolean"}
if use_anus:
  !wget https://civitai.com/api/download/models/1352390?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Clitanari Concept; futanari without balls
#@markdown [公式ページ](https://civitai.com/models/1236959/clitanari-concept-futanari-without-balls)
use_noballfuta = True #@param {type: "boolean"}
if use_noballfuta:
  !wget https://civitai.com/api/download/models/1394023?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/


#@markdown ### better-amateur-anal-lora-ponyxl
#@markdown [公式ページ](https://civitai.com/models/963175/better-amateur-anal-lora-ponyxl)
use_analbetter = False #@param {type: "boolean"}
if use_analbetter:
  !wget https://civitai.com/api/download/models/1078360?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### ディープスロート
#@markdown [公式ページ](https://civitai.com/models/2535814/deepthroat-slider?modelVersionId=2849922)
use_deepthroat = True #@param {type: "boolean"}
if use_deepthroat:
  !wget https://civitai.com/api/download/models/2849922?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### 奥深くまで挿入
#@markdown [公式ページ](https://civitai.com/models/550870/balls-deep-deeper-penetration?modelVersionId=1226551)
use_ballsdeep = True #@param {type: "boolean"}
if use_ballsdeep:
  !wget https://civitai.com/api/download/models/1226551?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/


#@markdown ### NoBalls-Futanari-PonYXL
#@markdown [公式ページ](https://civitai.com/models/492611/noballs-futanari-ponyxl)
use_NoBalls = False #@param {type: "boolean"}
if use_NoBalls:
  !wget https://civitai.com/api/download/models/547710?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Realism Lora By Stable Yogi (Pony)
#@markdown [公式ページ](https://civitai.com/models/1098033?modelVersionId=1233415)
use_Realism = False #@param {type: "boolean"}
if use_Realism:
  !wget https://civitai.com/api/download/models/1233415?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/



#@markdown ### Imouto (妹) - Imouto! Life ~Fantasy~
#@markdown [公式ページ](https://civitai.com/models/461085?modelVersionId=513143)
use_imout = False #@param {type: "boolean"}
if use_imout:
  !wget https://civitai.com/api/download/models/513143?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/






#@markdown ---
#@markdown ## LoRA (SDXL)

#@markdown ### XL1マンこ
use_XLpussy = False #@param {type: "boolean"}
if use_XLpussy:
  !wget https://civitai.com/api/download/models/232190?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### 着物(XL)
#@markdown [公式ページ](https://civitai.com/models/577696/kimono)
use_kimonoxl = False #@param {type: "boolean"}
if use_kimonoxl:
   !wget https://civitai.com/api/download/models/644190?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Background Aerial Landscape Scenery(XL)
#@markdown [公式ページ](https://civitai.com/models/292821/background-aerial-landscape-scenery)
use_haikeixl = False #@param {type: "boolean"}
if use_haikeixl:
   !wget https://civitai.com/api/download/models/329122?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### 神社 Jinja / shrine SDXL(XL)
#@markdown [公式ページ](https://civitai.com/models/232521/jinja-shrine-sdxl)
use_shrinexl = False #@param {type: "boolean"}
if use_shrinexl:
   !wget https://civitai.com/api/download/models/262452?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### 和室 washitsu SDXL(XL)
#@markdown [公式ページ](https://civitai.com/models/134682/washitsu-sdxl)
use_shrinexl = False #@param {type: "boolean"}
if use_shrinexl:
   !wget https://civitai.com/api/download/models/148378?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/




#@markdown ---
#@markdown ## LoRA (補助)

#@markdown ### スマホで電話
#@markdown [公式ページ](https://civitai.com/models/176948/talking-on-phone)
use_smartphone = False #@param {type: "boolean"}
if use_smartphone:
   !wget https://civitai.com/api/download/models/198638?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### 和室
#@markdown [公式ページ](https://civitai.com/models/47379/modern-genkan-in-japanese-houses)
use_japanese_room = False #@param {type: "boolean"}
if use_japanese_room:
   !wget https://civitai.com/api/download/models/201447?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### 正座
#@markdown [公式ページ](https://civitai.com/models/135506/comic-expression-white-eyes-set-a-a)
use_seiza = False #@param {type: "boolean"}
if use_seiza:
   !wget https://civitai.com/api/download/models/382458?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### 変顔
#@markdown [公式ページ](https://civitai.com/models/135506/comic-expression-white-eyes-set-a-a)
use_hengao = False #@param {type: "boolean"}
if use_hengao:
   !wget https://civitai.com/api/download/models/85488?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### 和服
#@markdown [公式ページ](https://civitai.com/models/119851/comic-expression-white-eyes-set-a-a)
use_wahuku = False #@param {type: "boolean"}
if use_wahuku:
   !wget https://civitai.com/api/download/models/130263?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### 日本背景
#@markdown [公式ページ](https://civitai.com/models/119851/comic-expression-white-eyes-set-a-a)
use_japanhaikei = False #@param {type: "boolean"}
if use_japanhaikei:
   !wget https://civitai.com/api/download/models/6046?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### 和服半脱ぎ
#@markdown [公式ページ](https://civitai.com/models/119851/comic-expression-white-eyes-set-a-a)
use_wahukunugi = False #@param {type: "boolean"}
if use_wahukunugi:
   !wget https://civitai.com/api/download/models/242195?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### 髪ブラ
#@markdown [公式ページ](https://civitai.com/models/910615/hair-censor)
use_datui = False #@param {type: "boolean"}
if use_datui:
   !wget https://civitai.com/api/download/models/1019054?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### 緊縛
#@markdown [公式ページ](https://civitai.com/models/119851/comic-expression-white-eyes-set-a-a)
use_kinbaku = False #@param {type: "boolean"}
if use_kinbaku:
   !wget https://civitai.com/api/download/models/249563?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### 美しい背景
#@markdown [公式ページ](https://civitai.com/models/119851/comic-expression-white-eyes-set-a-a)
use_bbg = False #@param {type: "boolean"}
if use_bbg:
   !wget https://civitai.com/api/download/models/570154?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### 神社
#@markdown [公式ページ](https://civitai.com/models/85660/jinja-shrine-sd15)
use_jinjya = False #@param {type: "boolean"}
if use_jinjya:
   !wget https://civitai.com/api/download/models/181725?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### 後ろから撮影
#@markdown [公式ページ](https://civitai.com/models/62448/back-view-perspective)
use_back = False #@param {type: "boolean"}
if use_back:
   !wget https://civitai.com/api/download/models/67000?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/




#@markdown ### 水ドレス
#@markdown [公式ページ](https://civitai.com/models/119851/comic-expression-white-eyes-set-a-a)
use_waterdress = False #@param {type: "boolean"}
if use_waterdress:
   !wget https://civitai.com/api/download/models/438950?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### 葉っぱビキニ
#@markdown [公式ページ](https://civitai.com/models/96688/zongyeunderwear)
use_waterdress = False #@param {type: "boolean"}
if use_waterdress:
   !wget https://civitai.com/api/download/models/563145?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### 黄金クロス水着
#@markdown [公式ページ](https://civitai.com/models/96688/zongyeunderwear)
use_waterdress = False #@param {type: "boolean"}
if use_waterdress:
   !wget https://civitai.com/api/download/models/431595?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### サンドドレス
#@markdown [公式ページ](https://civitai.com/models/267536/sand-maxi-dress)
use_sanddress = False #@param {type: "boolean"}
if use_sanddress:
   !wget https://civitai.com/api/download/models/301655?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### 空港
#@markdown [公式ページ](https://civitai.com/models/540564/airport-departures-lobby)
use_airport = False #@param {type: "boolean"}
if use_airport:
   !wget https://civitai.com/api/download/models/600980?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### ハローウィン
#@markdown [公式ページ](https://civitai.com/models/181461/halloween-by-stable-yogi)
use_airport = False #@param {type: "boolean"}
if use_airport:
   !wget https://civitai.com/api/download/models/203654?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### サンタドレス
#@markdown [公式ページ](https://civitai.com/models/56116/santa-outfit)
use_santa = False #@param {type: "boolean"}
if use_santa:
   !wget https://civitai.com/api/download/models/60525?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### 初詣
#@markdown [公式ページ](https://civitai.com/models/1057137/hrine-bell-hatsumoude)
use_hatsumoude = False #@param {type: "boolean"}
if use_hatsumoude:
   !wget https://civitai.com/api/download/models/1186249?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### コスプレ・進撃の巨人・ミサカ
#@markdown [公式ページ](https://civitai.com/models/196925/cosplaymikasa-ackerman-bondage-outfit-or-attack-on-titan)
use_cosplay = False #@param {type: "boolean"}
if use_cosplay:
   !wget https://civitai.com/api/download/models/221427?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### コスプレ・コードギアス・CC
#@markdown [公式ページ](https://civitai.com/models/18681/code-geass-cc-straitjacket-cosplay)
use_cosplay1 = False #@param {type: "boolean"}
if use_cosplay1:
   !wget https://civitai.com/api/download/models/24151?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### コスプレ・うる星奴等・ラムちゃん
#@markdown [公式ページ](https://civitai.com/models/65850/cosplay-outfit-or-lum-urusei-yatsura)
use_cosplay2 = False #@param {type: "boolean"}
if use_cosplay2:
   !wget https://civitai.com/api/download/models/70501?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/


#@markdown ### コスプレ・One Piece・ナミ
#@markdown [公式ページ](https://civitai.com/models/143243/cosplaynami-or-one-piece-east-blue)
use_cosplay3 = False #@param {type: "boolean"}
if use_cosplay3:
   !wget https://civitai.com/api/download/models/158938?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### コスプレ・フリーレン
#@markdown [公式ページ](https://civitai.com/models/275791/cosplayfrieren-bikini-modified-ver-or-frieren-beyond-journeys-end)
use_cosplay4 = False #@param {type: "boolean"}
if use_cosplay4:
   !wget https://civitai.com/api/download/models/310848?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/




#@markdown ---
#@markdown ### 商用不可lora

#@markdown ### Yellow swimsuit(水着)
#@markdown [公式ページ](https://civitai.com/models/138433/yellow-september)
use_yellow = False #@param {type: "boolean"}
if use_yellow:
   !wget https://civitai.com/api/download/models/152955?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### Slingshot Swimsuit(水着)
#@markdown [公式ページ](https://civitai.com/models/197877/slingshot-swimsuit-by-stable-yogi)
use_slingshot = False #@param {type: "boolean"}
if use_slingshot:
   !wget https://civitai.com/api/download/models/222591?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### おしゃれ1
#@markdown [公式ページ](https://civitai.com/models/87245/muggle-loraopen-coat)
use_osa1 = False #@param {type: "boolean"}
if use_osa1:
   !wget https://civitai.com/api/download/models/92843?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/




#@markdown ###  Thicker Lines Anime Style LoRA Mix
#@markdown [公式ページ](https://civitai.com/models/13910/thicker-lines-anime-style-lora-mix?modelVersionId=16368)
use_Thicker = False #@param {type: "boolean"}
if use_Thicker:
   !wget https://civitai.com/api/download/models/16368?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/


#@markdown ---
#@markdown ## LoRA (Base: SDXL)

#@markdown ### Latent Consistency Model (LCM) LoRA: SDXL
#@markdown [公式ページ](https://huggingface.co/latent-consistency/lcm-lora-sdxl)
use_lcm_lora_sdxl = False #@param {type: "boolean"}
if use_lcm_lora_sdxl:
  !wget https://huggingface.co/latent-consistency/lcm-lora-sdxl/resolve/main/pytorch_lora_weights.safetensors --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### SDXL-Lightning LoRA (2step)
#@markdown [公式ページ](https://huggingface.co/ByteDance/SDXL-Lightning)
use_sdxl_lightning_lora_2step = False #@param {type: "boolean"}
if use_sdxl_lightning_lora_2step:
  !wget https://huggingface.co/ByteDance/SDXL-Lightning/resolve/main/sdxl_lightning_2step_lora.safetensors --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### SDXL-Lightning LoRA (4step)
#@markdown [公式ページ](https://huggingface.co/ByteDance/SDXL-Lightning)
use_sdxl_lightning_lora_4step = False #@param {type: "boolean"}
if use_sdxl_lightning_lora_4step:
  !wget https://huggingface.co/ByteDance/SDXL-Lightning/resolve/main/sdxl_lightning_4step_lora.safetensors --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ### SDXL-Lightning LoRA (8step)
#@markdown [公式ページ](https://huggingface.co/ByteDance/SDXL-Lightning)
use_sdxl_lightning_lora_8step = False #@param {type: "boolean"}
if use_sdxl_lightning_lora_8step:
  !wget https://huggingface.co/ByteDance/SDXL-Lightning/resolve/main/sdxl_lightning_8step_lora.safetensors --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/models/Lora/

#@markdown ---
#@markdown ## VAE
#@markdown 導入したいVAEにチェックを入れてください

#@markdown ### vae-ft-mse-840000
#@markdown [公式ページ](https://huggingface.co/stabilityai/sd-vae-ft-mse-original)
use_vae_ft_mse_840000 = False #@param {type: "boolean"}
if use_vae_ft_mse_840000:
  !wget https://huggingface.co/stabilityai/sd-vae-ft-mse-original/resolve/main/vae-ft-mse-840000-ema-pruned.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/VAE/

#@markdown ### kl-f8-anime2
#@markdown [公式ページ](https://huggingface.co/hakurei/waifu-diffusion-v1-4)
use_klf8_anime2 = False #@param {type: "boolean"}
if use_klf8_anime2:
  !wget https://huggingface.co/hakurei/waifu-diffusion-v1-4/resolve/main/vae/kl-f8-anime2.ckpt --directory-prefix=/content/stable-diffusion-webui-reForge/models/VAE/

#@markdown ### SDXL - VAE (SDXL向け)
#@markdown [公式ページ](https://huggingface.co/stabilityai/sdxl-vae)
use_sdxl_vae = False #@param {type: "boolean"}
if use_sdxl_vae:
  !wget https://huggingface.co/stabilityai/sdxl-vae/resolve/main/sdxl_vae.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/models/VAE/

#@markdown Pony Standard VAE
#@markdown [公式ページ](https://civitai.com/models/660673/pony-standard-vae)
use_ponyvae = False #@param {type: "boolean"}
if use_ponyvae:
  !wget https://civitai.com/api/download/models/785437?token={civitai_api_key} --directory-prefix=/content/stable-diffusion-webui-reForge/models/VAE/


#@markdown ---
#@markdown ## embeddings
#@markdown 導入したいembeddingsにチェックを入れてください

#@markdown ### EasyNegative
#@markdown [公式ページ](https://huggingface.co/datasets/gsdf/EasyNegative)
use_easy_negative = False #@param {type: "boolean"}
if use_easy_negative:
  !wget https://huggingface.co/datasets/gsdf/EasyNegative/resolve/main/EasyNegative.safetensors --directory-prefix=/content/stable-diffusion-webui-reForge/embeddings/

#@markdown ### negative_hand Negative Embedding
#@markdown [公式ページ](https://civitai.com/models/56519/negativehand-negative-embedding)
use_negative_hand_negative_embedding = False #@param {type: "boolean"}
if use_negative_hand_negative_embedding:
  !wget https://civitai.com/api/download/models/60938?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/embeddings/

#@markdown ### badhandv4
#@markdown [公式ページ](https://civitai.com/models/16993/badhandv4-animeillustdiffusion)
use_badhandv4 = False #@param {type: "boolean"}
if use_badhandv4:
  !wget https://civitai.com/api/download/models/20068?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/embeddings/

#@markdown ### nagativeXL_D
#@markdown [公式ページ](https://civitai.com/models/118418?modelVersionId=134583)
use_nagativeXL_D = True #@param {type: "boolean"}
if use_nagativeXL_D:
  !wget https://civitai.com/api/download/models/134583?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/embeddings/

#@markdown ### Positive Embeddings By Stable Yogi
#@markdown [公式ページ](https://civitai.com/models/177792/negembedstableyogi)
use_positive = False #@param {type: "boolean"}
if use_positive:
  !wget https://civitai.com/api/download/models/775151?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/embeddings/

#@markdown ### negative Embeddings By Stable Yogi
#@markdown [公式ページ](https://civitai.com/models/177792/negembedstableyogi)
use_nagative = False #@param {type: "boolean"}
if use_nagative:
  !wget https://civitai.com/api/download/models/772342?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/embeddings/



#@markdown ### unaestheticXL | Negative TI(_Jug6)
#@markdown [公式ページ](https://civitai.com/models/119032/unaestheticxl-or-negative-ti)
use_unaesthetic_xl_jug6 = False #@param {type: "boolean"}
if use_unaesthetic_xl_jug6 and not use_blue_pencil_xl:
  !wget https://civitai.com/api/download/models/207934?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/embeddings/

#@markdown ### unaestheticXL | Negative TI(_Sky3.1)
#@markdown [公式ページ](https://civitai.com/models/119032/unaestheticxl-or-negative-ti)
use_unaesthetic_xl_sky31 = False #@param {type: "boolean"}
if use_unaesthetic_xl_sky31 and not use_blue_pencil_xl:
  !wget https://civitai.com/api/download/models/175819?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/embeddings/

#@markdown ### unaestheticXL | Negative TI(_AYv1)
#@markdown [公式ページ](https://civitai.com/models/119032/unaestheticxl-or-negative-ti)
use_unaesthetic_xl_ayv1 = False #@param {type: "boolean"}
if use_unaesthetic_xl_ayv1 and not use_blue_pencil_xl:
  !wget https://civitai.com/api/download/models/162146?token={civitai_api_key} --content-disposition --directory-prefix=/content/stable-diffusion-webui-reForge/embeddings/

#@markdown ---
#@markdown ## Upscaler

#@markdown ### 4x-UltraSharp

#@markdown [公式ページ](https://openmodeldb.info/models/4x-UltraSharp)

#@markdown #### ダウンロードについて

#@markdown こちらはMegaにファイルがホストされている関係でプログラム側から直接ダウンロードが出来ません。[4x-UltraSharp](https://openmodeldb.info/models/4x-UltraSharp)より直接ダウンロードした上で、/content/stable-diffusion-webui-reForge/models/ESRGANにファイルを格納してご利用ください。


#@markdown ### 4x-AnimeSharp

#@markdown [公式ページ](https://openmodeldb.info/models/4x-AnimeSharp)

#@markdown #### ダウンロードについて

#@markdown こちらはMegaにファイルがホストされている関係でプログラム側から直接ダウンロードが出来ません。[4x-AnimeSharp](https://openmodeldb.info/models/4x-AnimeSharp)より直接ダウンロードした上で、/content/stable-diffusion-webui-reForge/models/ESRGANにファイルを格納してご利用ください。


#@markdown ---
#@markdown ## 拡張機能
#@markdown 導入したい拡張機能にチェックを入れてください

#@markdown ### After Detailer
#@markdown [公式ページ](https://github.com/Bing-su/adetailer)

#@markdown [使い方を確認する(YouTube)](https://youtu.be/WJfhNjzjMIE)
use_after_detailer = True #@param {type: "boolean"}
if use_after_detailer:
  !git clone https://github.com/Bing-su/adetailer.git /content/stable-diffusion-webui-reForge/extensions/adetailer



#@markdown ### Rembg
#@markdown [公式ページ](https://github.com/AUTOMATIC1111/stable-diffusion-webui-rembg)

#@markdown [使い方を確認する(YouTube)](https://youtu.be/nK9M7NYgZDI)
use_rembg = False #@param {type: "boolean"}
if use_rembg:
  !git clone https://github.com/AUTOMATIC1111/stable-diffusion-webui-rembg.git /content/stable-diffusion-webui-reForge/extensions/stable-diffusion-webui-rembg

#@markdown ### sd-webui-enable-checker
#@markdown [公式ページ](https://github.com/shirayu/sd-webui-enable-checker)

#@markdown [使い方を確認する(YouTube)](https://youtu.be/e_LTlWbKTaE)
use_sd_webui_enable_chacker = False #@param {type: "boolean"}
if use_sd_webui_enable_chacker:
  !git clone https://github.com/shirayu/sd-webui-enable-checker.git /content/stable-diffusion-webui-reForge/extensions/sd-webui-enable-checker

#@markdown ### Easy Generate Forever
#@markdown [公式ページ](https://github.com/blue-pen5805/sdweb-easy-generate-forever)

use_easy_generate_forever = False  #@param {type: "boolean"}
if use_easy_generate_forever:
  !git clone https://github.com/blue-pen5805/sdweb-easy-generate-forever.git /content/stable-diffusion-webui-reForge/extensions/sdweb-easy-generate-forever


#@markdown ### sd-forge-couple
#@markdown [公式ページ](https://github.com/Haoming02/sd-forge-couple)

use_sdforgecouple = True  #@param {type: "boolean"}
if use_sdforgecouple:
  !git clone https://github.com/Haoming02/sd-forge-couple.git /content/stable-diffusion-webui-reForge/extensions/sd-forge-couple

#@markdown ### sd-Dynamic Prompts
#@markdown [公式ページ](https://github.com/Haoming02/sd-forge-couple)

use_Dynamic = True  #@param {type: "boolean"}
if use_Dynamic:
  !git clone https://github.com/adieyal/sd-dynamic-prompts.git /content/stable-diffusion-webui-reForge/extensions/sd-dynamic-prompts

#@markdown ### Inpaint Anything for Stable Diffusion Web UI
#@markdown [公式ページ](https://github.com/Uminosachi/sd-webui-inpaint-anything)
use_sd_webui_inpaint_anything = False #@param {type: "boolean"}
if use_sd_webui_inpaint_anything:
  # Inpaint Anything for Stable Diffusion Web UI
  !git clone https://github.com/Uminosachi/sd-webui-inpaint-anything.git /content/stable-diffusion-webui-reForge/extensions/sd-webui-inpaint-anything

#@markdown ### Segment Anything for Stable Diffusion WebUI
#@markdown [公式ページ](https://github.com/continue-revolution/sd-webui-segment-anything)
use_sd_webui_segment_anything = False #@param {type: "boolean"}
if use_sd_webui_segment_anything:
  # Segment Anything for Stable Diffusion WebUI
  !git clone https://github.com/continue-revolution/sd-webui-segment-anything.git /content/stable-diffusion-webui-reForge/extensions/sd-webui-segment-anything
  # 必要なモデルファイルをダウンロード - Meta AI (Colab利用前提のためhuge, largeのみダウンロード)
  !wget https://dl.fbaipublicfiles.com/segment_anything/sam_vit_h_4b8939.pth --directory-prefix=/content/stable-diffusion-webui-reForge/extensions/sd-webui-segment-anything/models/sam
  !wget https://dl.fbaipublicfiles.com/segment_anything/sam_vit_l_0b3195.pth --directory-prefix=/content/stable-diffusion-webui-reForge/extensions/sd-webui-segment-anything/models/sam


# WebUI を起動

#@markdown ---
#@markdown ## 起動
# 必ず WEBUI_DIR に移動してから launch（パスずれ対策）
import os
os.chdir("/content/stable-diffusion-webui-reForge")
print("cwd:", os.getcwd())
assert os.path.isfile("launch.py"), f"launch.py not found in {os.getcwd()}"

# Colab の TensorFlow 2.20 は protobuf 5.28 以上でないと import できない。
# reForge は protobuf を 4 系以下に戻す。transformers は TensorFlow が見えると
# CLIP 読み込みでそれを import し、runtime_version で落ちる。WebUI は PyTorch だけ使う。
# Colab の wandb 0.28 は pydantic v2 の AliasChoices が必要で、
# reForge の pydantic 1.10.15 と衝突する。Lightning は起動時に wandb を読むので、その import だけ外す。
os.environ["USE_TF"] = "0"
os.environ["USE_TORCH"] = "1"
os.environ["USE_FLAX"] = "0"
!pip uninstall -y -q tensorflow tensorflow-cpu tensorflow-gpu tf-keras keras
starter = r"""
import os
os.environ["USE_TF"] = "0"
os.environ["USE_TORCH"] = "1"
os.environ["USE_FLAX"] = "0"

def _skip_wandb_logger():
    import sysconfig
    from pathlib import Path
    roots = [Path(sysconfig.get_paths()["purelib"]), Path(sysconfig.get_paths()["platlib"])]
    found = []
    for root in roots:
        found.extend(root.glob("pytorch_lightning/loggers/__init__.py"))
    if not found:
        raise SystemExit("pytorch_lightning loggers/__init__.py not found")
    init_py = found[0]
    text = init_py.read_text(encoding="utf-8")
    needle = "from pytorch_lightning.loggers.wandb import WandbLogger"
    if needle not in text:
        print("wandb logger import already skipped:", init_py)
        return
    out = []
    for line in text.splitlines(True):
        if needle in line:
            out.append("WandbLogger = None  # Colab wandb needs pydantic v2; reForge pins pydantic 1.10\n")
        else:
            out.append(line)
    init_py.write_text("".join(out), encoding="utf-8")
    print("skipped WandbLogger import in", init_py)

import launch
_real_prepare = launch.prepare_environment

def _prepare():
    _real_prepare()
    _skip_wandb_logger()

launch.prepare_environment = _prepare
launch.main()
"""
with open("colab_launch.py", "w", encoding="utf-8") as f:
    f.write(starter)
# 工場リンク。この回の公開アドレスを MyDrive/sd-diff-set/colab-link.json に書く。
# 同じセルの cmd_args を使う。Civitai のキーとログインは Secrets から読み、表示しない。

import base64
import binascii
import io
import json
import os
import re
import shutil
import subprocess
import sys
import threading
import time
import urllib.error
import urllib.request
import zipfile
from datetime import datetime, timezone
from pathlib import Path

NOTEBOOK_ID = "1U92UAuSNJ3OHfgM5KTjA9maDUlJP0ySf"
_URL = re.compile(r"https://[a-z0-9-]+\.gradio\.live", re.I)
_FACTORY_B64 = ""  # bundle
_FACTORY_ZIP_ID = "1N998IDTfUM8KA0U68RYkG-A7t1Pf005H"  # zip-id


def _link_path() -> Path:
    shared = os.environ.get("SD_DIFF_SHARED_DRIVE", "").strip()
    root = Path("/content/gdrive")
    if shared and (root / "Shareddrives" / shared).is_dir():
        base = root / "Shareddrives" / shared
    else:
        base = root / "MyDrive"
    dest = base / "sd-diff-set"
    dest.mkdir(parents=True, exist_ok=True)
    return dest / "colab-link.json"


def _write(origin: str) -> None:
    payload = {
        "api": origin,
        "notebook_id": NOTEBOOK_ID,
        "written_at": datetime.now(timezone.utc).strftime("%Y-%m-%dT%H:%M:%SZ"),
    }
    path = _link_path()
    path.write_text(json.dumps(payload, ensure_ascii=False, indent=2) + "\n", encoding="utf-8")
    print("工場リンクを書きました:", path)


def _output_root() -> Path:
    root = _link_path().parent / "outputs"
    root.mkdir(parents=True, exist_ok=True)
    return root


def _sync_outputs(port: str, pointed: list[bool]) -> int:
    """Copy any PNG still sitting on the VM. Each save also copies itself."""
    del port
    ext = Path("/content/stable-diffusion-webui-reForge/extensions/sd-diff-set")
    if ext.is_dir() and str(ext) not in sys.path:
        sys.path.insert(0, str(ext))
    from sd_diff_set.drive_save import sync_local_outputs

    copied, root = sync_local_outputs()
    if root is not None and not pointed[0]:
        pointed[0] = True
        print("出力先:", root)
    if copied:
        print("ドライブへ:", copied)
    return copied


def _output_line(proc: subprocess.Popen[str], port: str) -> None:
    pointed = [False]
    while proc.poll() is None:
        try:
            _sync_outputs(port, pointed)
        except (OSError, urllib.error.URLError) as exc:
            print("出力の同期に失敗:", exc)
        time.sleep(10)


def _download_factory_zip(file_id: str) -> bytes:
    url = "https://drive.google.com/uc?export=download&id=" + file_id
    request = urllib.request.Request(url, headers={"User-Agent": "sd-diff-set"})
    try:
        with urllib.request.urlopen(request, timeout=120) as response:
            data = response.read()
    except (OSError, urllib.error.URLError) as exc:
        print("工場のzipを取れませんでした:", type(exc).__name__)
        return b""
    if data[:2] != b"PK":
        print("工場のzipを取れませんでした")
        return b""
    return data


def _bundled_extension() -> Path | None:
    """Panel zip from Drive. A legacy inline payload still works, but the notebook no longer carries it."""
    payload = _FACTORY_B64.strip()
    raw = b""
    if payload:
        try:
            raw = base64.b64decode(payload.encode("ascii"), validate=True)
        except (ValueError, binascii.Error):
            print("工場の同梱データを読めませんでした")
            return None
    elif _FACTORY_ZIP_ID.strip():
        raw = _download_factory_zip(_FACTORY_ZIP_ID.strip())
        if not raw:
            return None
    else:
        return None
    try:
        with zipfile.ZipFile(io.BytesIO(raw)) as archive:
            root = Path("/tmp/sd-diff-set-bundled")
            if root.exists():
                shutil.rmtree(root)
            root.mkdir(parents=True)
            archive.extractall(root)
    except (OSError, ValueError, binascii.Error, zipfile.BadZipFile):
        print("工場の同梱データを読めませんでした")
        return None
    bundled = root / "extension"
    if not (bundled / "sd_diff_set" / "ui.py").is_file():
        print("工場の同梱データに操作盤がありません")
        return None
    return bundled


_DETECTOR_WEIGHTS = (
    ("face_yolov8n.pt", 6230011, "https://huggingface.co/Bingsu/adetailer/resolve/main/face_yolov8n.pt", False),
    ("hand_yolov8n.pt", 6237883, "https://huggingface.co/Bingsu/adetailer/resolve/main/hand_yolov8n.pt", False),
    ("person_yolov8n-seg.pt", 6777003, "https://huggingface.co/Bingsu/adetailer/resolve/main/person_yolov8n-seg.pt", False),
    ("foot_yolov8x_v2.pt", 136712062, "https://huggingface.co/MonetEinsley/ADetailer_CM/resolve/main/foot_yolov8x_v2.pt", False),
    ("nsfw-anime-medium-x1280.pt", 54575409, "https://civitai.com/api/download/models/2941061", True),
)


def _weight_ok(path: Path, size: int) -> bool:
    try:
        return path.is_file() and path.stat().st_size == size
    except OSError:
        return False


def _fetch_detector_weights(model_dir: Path | None) -> None:
    """Copy Drive weights, then fetch a missing file once. The factory itself still does not download."""
    folder = _link_path().parent / "weights"
    try:
        folder.mkdir(parents=True, exist_ok=True)
    except OSError as exc:
        print("重みのフォルダを作れませんでした:", type(exc).__name__)
        return
    if model_dir is not None:
        model_dir.mkdir(parents=True, exist_ok=True)
    key = _secret("CIVITAI_API_KEY")
    for name, size, url, needs_key in _DETECTOR_WEIGHTS:
        placed = [folder / name]
        if model_dir is not None:
            placed.insert(0, model_dir / name)
        ready = [path for path in placed if _weight_ok(path, size)]
        if ready:
            for dest in placed:
                if any(dest.resolve() == found.resolve() for found in ready):
                    continue
                try:
                    shutil.copy2(ready[0], dest)
                except OSError as exc:
                    print("重みのコピーを飛ばしました:", name, type(exc).__name__)
            continue
        if needs_key and not key:
            print("重みを取っていません:", name, "CIVITAI_API_KEY がありません")
            continue
        headers = {"User-Agent": "sd-diff-set"}
        if needs_key:
            headers["Authorization"] = "Bearer " + key
        temporary = folder / (name + ".part")
        try:
            request = urllib.request.Request(url, headers=headers)
            with urllib.request.urlopen(request, timeout=600) as response, temporary.open("wb") as out:
                while True:
                    chunk = response.read(1024 * 1024)
                    if not chunk:
                        break
                    out.write(chunk)
                    out.flush()
        except (OSError, urllib.error.URLError) as exc:
            print("重みを取れませんでした:", name, type(exc).__name__)
            temporary.unlink(missing_ok=True)
            continue
        if not _weight_ok(temporary, size):
            print("重みの大きさが違います:", name)
            temporary.unlink(missing_ok=True)
            continue
        saved = folder / name
        temporary.replace(saved)
        if model_dir is not None:
            try:
                shutil.copy2(saved, model_dir / name)
            except OSError as exc:
                print("重みのコピーを飛ばしました:", name, type(exc).__name__)
        print("重みを入れました:", name)


# 起動前にマイドライブから置く。行頭が # のパスは動かさない。
_DRIVE_COPIES = (
    ("/content/gdrive/MyDrive/controlnet/diffusers_xl_depth_full.safetensors", "models/ControlNet"),
    # /content/gdrive/MyDrive/ComfyUI/stable-diffusion-webui/extensions/sd-webui-controlnet/models/t2i-adapter_diffusers_xl_depth_midas.safetensors models/ControlNet
    # /content/gdrive/MyDrive/ComfyUI/stable-diffusion-webui/extensions/sd-webui-controlnet/models/thibaud_xl_openpose.safetensors models/ControlNet
    ("/content/gdrive/MyDrive/controlnet/diffusers_xl_canny_full.safetensors", "models/ControlNet"),
    # /content/gdrive/MyDrive/controlnet/CN-anytest_v4-marged.safetensors models/ControlNet
    # /content/gdrive/MyDrive/controlnet/config.json models/ControlNet
    # /content/gdrive/MyDrive/controlnet/diffusers_xl_depth_mid.safetensors models/ControlNet
    # /content/gdrive/MyDrive/controlnet/pooXL28Pony29.safetensors models/ControlNet
    # /content/gdrive/MyDrive/ComfyUI/models/loras/unchi_pony_v1.safetensors models/Lora
    ("/content/gdrive/MyDrive/controlnet/DefecationIllustV1.safetensors", "models/Lora"),
    ("/content/gdrive/MyDrive/controlnet/caststyle_xl_epoch08.safetensors", "models/Lora"),
    ("/content/gdrive/MyDrive/controlnet/sora_xl_epoch08.safetensors", "models/Lora"),
    # /content/gdrive/MyDrive/ComfyUI/models/loras/tentaclejob_onaho_v1_pony.safetensors models/Lora
    ("/content/gdrive/MyDrive/controlnet/Eyes.pt", "models/adetailer"),
    # /content/gdrive/MyDrive/controlnet/ponyStandardVAE_v10.safetensors models/VAE
    ("/content/gdrive/MyDrive/Loras/MyLoRA/output/my_lora_fixed_for_webui.safetensors", "models/Lora"),
)


def _copy_drive_models(webui: Path) -> None:
    """Copy the listed Drive files before launch. A missing file does not stop the cell."""
    if not webui.is_dir():
        print("reForge がまだ無いので、追加のモデルはコピーしていません")
        return
    for src_text, folder in _DRIVE_COPIES:
        src = Path(src_text)
        dest_dir = webui / folder
        try:
            dest_dir.mkdir(parents=True, exist_ok=True)
            if not src.is_file():
                print("コピーを飛ばしました:", src.name)
                continue
            dest = dest_dir / src.name
            if dest.is_file() and dest.stat().st_size == src.stat().st_size:
                print("コピー済み:", src.name)
                continue
            shutil.copy2(src, dest)
        except OSError as exc:
            print("コピーを飛ばしました:", src.name, type(exc).__name__)
            continue
        print("コピーしました:", src.name)


def _install_factory() -> None:
    webui = Path("/content/stable-diffusion-webui-reForge")
    _copy_drive_models(webui)
    model_dir = webui / "models" / "adetailer" if webui.is_dir() else None
    _fetch_detector_weights(model_dir)
    if not webui.is_dir():
        print("reForge がまだ無いので、工場は入れていません")
        return
    drive_dir = _link_path().parent / "extension"
    src = _bundled_extension()
    if src is None and (drive_dir / "sd_diff_set" / "ui.py").is_file():
        src = drive_dir
    if src is None:
        print("工場の拡張はまだありません:", drive_dir)
        return
    if src != drive_dir:
        try:
            if drive_dir.exists():
                shutil.rmtree(drive_dir)
            shutil.copytree(src, drive_dir, ignore=shutil.ignore_patterns("__pycache__", "*.pyc"))
        except OSError as exc:
            print("ドライブへの工場コピーは飛ばしました:", type(exc).__name__)
    dest = webui / "extensions" / "sd-diff-set"
    if dest.exists():
        shutil.rmtree(dest)
    shutil.copytree(src, dest, ignore=shutil.ignore_patterns("__pycache__", "*.pyc"))
    print("工場を入れました:", dest)


def _launch_script() -> str:
    if Path("colab_launch.py").is_file():
        return "colab_launch.py"
    return "launch.py"


def _pin_detector_imports() -> None:
    """Keep the ADetailer installer from resolving ultralytics 8.4 and numpy 2."""
    install = Path("/content/stable-diffusion-webui-reForge/extensions/adetailer/install.py")
    needle = '("ultralytics", "8.3.75", None)'
    pinned = '("ultralytics", "8.3.75", "8.3.75")'
    if install.is_file():
        text = install.read_text(encoding="utf-8")
        if needle in text:
            install.write_text(text.replace(needle, pinned, 1), encoding="utf-8")
            print("検出器の版を固定しました: ultralytics==8.3.75")
        elif pinned in text:
            print("検出器の版は固定済みです: ultralytics==8.3.75")
        else:
            print("検出器の版を固定できませんでした: install.py の行が違います")
    else:
        print("ADetailer がまだ無いので、検出器の版は固定していません")
    try:
        completed = subprocess.run(
            [
                sys.executable,
                "-m",
                "pip",
                "install",
                "ultralytics==8.3.75",
                "numpy>=1.26.4,<2",
            ],
            check=False,
        )
    except OSError as exc:
        print("検出器の導入を飛ばしました:", type(exc).__name__)
        return
    if completed.returncode != 0:
        print("検出器の導入に失敗しました:", completed.returncode)


def _probe_detectors() -> None:
    """Import numpy, cv2, and YOLO once with the Python that starts webui."""
    code = "\n".join(
        [
            "import numpy, cv2",
            "from ultralytics import YOLO",
            "print('検出器を読めました: numpy', numpy.__version__)",
        ]
    )
    try:
        result = subprocess.run(
            [sys.executable, "-c", code],
            capture_output=True,
            text=True,
            timeout=180,
        )
    except (OSError, subprocess.TimeoutExpired) as exc:
        print("検出器の確認に失敗しました:", type(exc).__name__)
        return
    if result.returncode == 0:
        lines = [line for line in result.stdout.splitlines() if line.startswith("検出器を読めました")]
        print(lines[-1] if lines else "検出器を読めました")
        return
    detail = ""
    for line in (result.stderr or "").splitlines():
        if line.strip():
            detail = line.strip()
    if not detail:
        detail = "code " + str(result.returncode)
    print("検出器の確認に失敗しました:", detail[:300])


def _secret(name: str) -> str:
    """Colab Secret, then the process environment. The value is not printed."""
    value = os.environ.get(name, "").strip()
    if value:
        return value
    try:
        from google.colab import userdata  # type: ignore
    except ImportError:
        return ""
    try:
        found = userdata.get(name)
    except Exception:
        return ""
    return str(found or "").strip()


def _show(line: str) -> str:
    if "--gradio-auth" in line or "CIVITAI_API_KEY=" in line:
        return ""
    if "token=" in line:
        return re.sub(r"(token=)[^&\s]+", r"\1", line)
    return line


_install_factory()
_pin_detector_imports()
_probe_detectors()
args = str(cmd_args).split()  # noqa: F821  notebook cell defines cmd_args
if "--api" not in args:
    args.append("--api")
auth = _secret("GRADIO_AUTH")
if ":" in auth and "--gradio-auth" not in args:
    args.extend(["--gradio-auth", auth])
env = os.environ.copy()
env["PYTHONUNBUFFERED"] = "1"
env.setdefault("USE_TF", "0")
env.setdefault("USE_TORCH", "1")
env.setdefault("USE_FLAX", "0")
api_key = _secret("CIVITAI_API_KEY")
if api_key:
    env["CIVITAI_API_KEY"] = api_key
port = "7860"
if "--port" in args:
    port_at = args.index("--port") + 1
    if port_at < len(args):
        port = args[port_at]

def _api_ready(check_port: str) -> bool:
    try:
        with urllib.request.urlopen("http://127.0.0.1:" + check_port + "/sdapi/v1/memory", timeout=5) as response:
            return int(getattr(response, "status", 200)) < 500
    except (OSError, urllib.error.URLError):
        return False

def _pump(proc: subprocess.Popen[str], check_port: str) -> None:
    synced = False
    assert proc.stdout is not None
    for line in proc.stdout:
        shown = _show(line)
        if shown:
            sys.stdout.write(shown)
            sys.stdout.flush()
        if synced:
            continue
        match = _URL.search(line)
        if match is None:
            continue
        try:
            _write(match.group(0))
        except OSError as exc:
            print("工場リンクを書けませんでした:", exc)
        print("スマホ:", match.group(0) + "/?__theme=dark&phone=1")
        threading.Thread(target=_output_line, args=(proc, check_port), daemon=True).start()
        synced = True

if _api_ready(port):
    print("reForge は起動済みです: http://127.0.0.1:" + port)
else:
    proc = subprocess.Popen(
        [sys.executable, "-u", _launch_script(), *args],
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1,
        env=env,
    )
    threading.Thread(target=_pump, args=(proc, port), daemon=True).start()
    deadline = time.time() + 45 * 60
    while time.time() < deadline:
        if _api_ready(port):
            print("localhost API: http://127.0.0.1:" + port)
            break
        if proc.poll() is not None:
            print("reForge が止まりました:", proc.returncode)
            break
        time.sleep(5)
    else:
        print("API はまだです: http://127.0.0.1:" + port)
````

````
import os, subprocess, glob, py_compile, datetime
from google.colab import userdata, drive
drive.mount('/content/drive')
tok = userdata.get('GH_READ_SD_DIFF_EXT')
COMMIT = '3a575128'
dst = '/content/sd-diff-set-ext'
res = []
try:
    subprocess.run(['rm','-rf',dst])
    r = subprocess.run(['git','clone','--no-checkout',f'https://x-access-token:{tok}@github.com/fireworker011/sd-diff-set-ext.git',dst],capture_output=True,text=True)
    res.append('clone rc=%d %s' % (r.returncode, (r.stderr or '').replace(tok,'***')[-500:]))
    r = subprocess.run(['git','-C',dst,'checkout',COMMIT],capture_output=True,text=True)
    res.append('checkout rc=%d %s' % (r.returncode, r.stderr[-300:]))
    subprocess.run(['git','-C',dst,'remote','set-url','origin','https://github.com/fireworker011/sd-diff-set-ext.git'])
finally:
    del tok
head = subprocess.run(['git','-C',dst,'rev-parse','HEAD'],capture_output=True,text=True).stdout.strip()
ok_head = head.startswith(COMMIT)
pyfiles = glob.glob(dst+'/**/*.py', recursive=True)
errs = []
for f in pyfiles:
    try: py_compile.compile(f, doraise=True)
    except Exception as e: errs.append(f'{f}: {e}')
tree = subprocess.run(['bash','-c',f'cd {dst} && ls -la && find . -maxdepth 2 -not -path "./.git*" | sort | head -80'],capture_output=True,text=True).stdout
out = '/content/drive/MyDrive/sd-diff-set/outputs/smoke-20261113'
os.makedirs(out, exist_ok=True)
with open(out+'/HANDOFF.md','w') as fh:
    fh.write(f"# smoke-20261113 (load-only, no image generation)\n\n- time(UTC): {datetime.datetime.utcnow().isoformat()}\n- HEAD: {head}\n- HEAD starts with {COMMIT}: {'PASS' if ok_head else 'FAIL'}\n- python files: {len(pyfiles)}, py_compile errors: {len(errs)} -> {'PASS' if ok_head and not errs else 'FAIL'}\n- reForge extension load in webui: 未実施 (CG生成の再開の言葉が無いため、webui起動・画像生成はしていない)\n- category routing display: 未実施 (同上)\n\n## log\n" + '\n'.join(res) + "\n\n## py_compile errors (full)\n" + ('\n'.join(errs) or 'none') + "\n\n## tree\n```\n" + tree + "\n```\n")
print('HEAD', head, 'ok' if ok_head else 'NG'); print('py', len(pyfiles), 'errors', len(errs)); print('\n'.join(errs[:20]))
````
