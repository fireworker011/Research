## 話 + 上から 7 つ + 病棟の追加（迷ったらそのまま）

**話** — どの予告を描くか

- **番台ショート（25秒）** … 25秒・ミッション失敗。理容室だけ新しく描く
- **番台フル（92秒）** … 初回の92秒版。全部描く

**1. つなぎ方** — 動画をどう繋げるか

- **カット（本ごと独立・迷ったらこれ）** … 1本目から全部 T2V。プロンプトで直したい。カメラも変えられる
- **前の最終フレームから続ける** … 1本目は T2V。2本目以降を前クリップの最後のコマから I2V。つながり優先
- **用意した最終フレームへ着く** … stills の jpg を最後のコマにする。jpg がある話だけ

**2. カメラ**

- **横スク（真横・全身・迷ったらこれ）** … 常に真横の 2D 横スクロール
- **3Dアクション（引きの三人称）** … 引きの三人称。カットつなぎのとき画角が回る

**3. 画質**

- **スピード（最速）** … 試し打ち。turbo 4step。格闘 LoRA なし
- **バランス（迷ったらこれ）** … 普段使い。この UNet では LoRA なし euler 8
- **質（きれい・時間かかる）** … きれい優先。この UNet では LoRA なし euler 12

**4. 格闘 LoRA** — ハイメモリ専用の任意

- **格闘LoRAオフ（迷ったらこれ）** … 行為ルート。戦いなし。後背位とベロチュー。A100 40GB はこれ
- **格闘LoRAオン（ハイメモリ専用）** … 戦いルート。06と10が横スク。話も切り替わる。ハイメモリ専用

**5. 構成** — 病棟の話。完了か失敗かがここで分かれる

- **○受け入れる（生存・完了・迷ったらこれ）** … 戸惑いながら身体を張って出口へ。ミッション完了
- **□誘う（淫欲・失敗）** … こちらから攻める濃厚セックス。院内で淫欲に呑まれて失敗
- **×回避（出口・完了）** … エロを避けて一人で出口。ミッション完了
- **△戦って勝つ（完了・ハイメモリ）** … れいを倒して出口。みきは回避。かなはすり抜け。しのを倒す。ハイメモリ
- **△戦って負ける（敗北H・失敗・ハイメモリ）** … しのに倒されて専用敗北H。ミッション失敗。ハイメモリ

**6. 誘うポーズ** — 病棟の □誘う だけ。霞東は無視

- **四つん這い股広げ（迷ったらこれ）** … その場で四つん這い、股を広げて誘う
- **M字開脚仰向け** … 仰向けでM字に開いて誘う
- **ベロチュー→じゅぼ→騎乗位** … 笑って抱きつき、ベロチュー、じゅぼ、押し倒して騎乗位

**7. トイレ** — 病棟の道中。どれを選んでも次のシーンへ。霞東は無視

- **トイレに行かない（迷ったらこれ）** … 道中は廊下のまま。次はれい
- **トイレ・小便** … 汚い多目的トイレの洋式に座って小便。どれでも次へ
- **トイレ・オナニー** … 同じ便座でオナニー。どれでも次へ
- **トイレ・触手** … 便器から出た触手がマンコと肛門。どれでも次へ

登場チェックを外すと、その感染者のシーンを飛ばす（みき / れい / かな / しの）。

**シーンごと（病棟）** — 誘う（誘い方含む）・受け入れる・回避。5番が戦いのときは無視。霞東は無視。最後に残った人の構成で完了／失敗が決まる。

- **全体に従う（迷ったらこれ）** … 5番の構成をこのシーンでも使う
- **○受け入れる** … この人だけ戸惑いながら身体を張る
- **□誘う・四つん這い股広げ** … この人を四つん這いで誘う
- **□誘う・M字開脚仰向け** … この人を仰向けM字で誘う
- **□誘う・ベロチュー→じゅぼ→騎乗位** … この人をじゅぼのあと乗せて誘う
- **×回避** … この人だけキスですり抜ける

シネマ LoRA は積まない。スローモーションの語は書かない。視点は三人称ゲームのまま。

- 本番の inbox / queued / output は触らない。`models/` だけ共有
